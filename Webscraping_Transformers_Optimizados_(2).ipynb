## **WebScraper Periodicos (General)**

```python
class ScraperPeriodico(ABC):
    """Clase base abs para webscrapping de periódicos"""

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str):
        """
        Args:
            termino: Término de búsqueda
            fecha_desde: Fecha inicio (YYYY-MM-DD)
            fecha_hasta: Fecha fin (YYYY-MM-DD)
        """
        self.termino = termino
        self.fecha_desde = fecha_desde
        self.fecha_hasta = fecha_hasta
        self.session = requests.Session()
        self.session.headers.update({
        })

    @property
    @abstractmethod
    def nombre_periodico(self) -> str:
        """Nombre del periódico"""
        pass

    @abstractmethod
    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        """Construye la URL de búsqueda para una página específica"""
        pass

    @abstractmethod
    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        """Obtiene el total de páginas de resultados"""
        pass

    @abstractmethod
    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """Extrae los links de artículos de una página"""
        pass

    def scrape(self) -> pd.DataFrame:
        """Método principal de scraping (común para todos)"""
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")

        # Obtener total de páginas
        try:
            url_inicial = self._construir_url_busqueda(pagina=1)
            r = self.session.get(url_inicial, timeout=10)
            soup = BeautifulSoup(r.content, 'html.parser')
            total_paginas = self._obtener_total_paginas(soup)
            print(f"Total de páginas: {total_paginas}")
        except Exception as e:
            print(f"Error obteniendo páginas: {e}")
            return pd.DataFrame()

        # Recolectar links de todas las páginas
        links = self._recolectar_links(total_paginas)

        if not links:
            print("No se encontraron links")
            return pd.DataFrame()

        # Descargar artículos
        df = self._descargar_articulos(links)
        return df

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        """Recolecta links de todas las páginas"""
        links = []

        for pagina in range(1, total_paginas + 1):
            if pagina % 5 == 0 or pagina == 1:
                print(f"Recolectando links: página {pagina}/{total_paginas}")

            try:
                url = self._construir_url_busqueda(pagina)
                r = self.session.get(url, timeout=10)
                soup = BeautifulSoup(r.content, 'html.parser')
                links_pagina = self._extraer_links_pagina(soup)
                links.extend(links_pagina)
                time.sleep(0.5)  # Rate limiting
            except Exception as e:
                print(f"Error en página {pagina}: {e}")
                continue

        # Eliminar duplicados
        links = list(dict.fromkeys(links))
        print(f"Total de links únicos: {len(links)}")
        return links

    def _descargar_articulo_individual(self, info: tuple) -> Optional[Dict]:
        """Descarga un artículo individual"""
        i, link = info
        try:
            article = Article(link)
            article.download()
            article.parse()

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": article.publish_date,
                "texto": article.text
            }
        except:
            return None

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        """Descarga artículos en paralelo"""
        print(f"\nDescargando {len(links)} artículos...")

        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]

        with ThreadPoolExecutor(max_workers=10) as executor:
            resultados = list(executor.map(self._descargar_articulo_individual, links_enumerados))

        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")

        return pd.DataFrame(data)
```

#### **El Diario**

```python
class ScraperElDiario(ScraperPeriodico):
    REGION = "eje_cafetero_antioquia"
    """Scraper específico para El Diario (Pereira)"""

    BASE_URL = "https://www.eldiario.com.co"

    MESES = {
        'enero': 1, 'febrero': 2, 'marzo': 3, 'abril': 4,
        'mayo': 5, 'junio': 6, 'julio': 7, 'agosto': 8,
        'septiembre': 9, 'octubre': 10, 'noviembre': 11, 'diciembre': 12
    }

    @property
    def nombre_periodico(self) -> str:
        return "El Diario"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        return f"{self.BASE_URL}/?s={self.termino}&paged={pagina}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        try:
            script_content = soup.find_all('script', string=re.compile('max_num_pages'))
            for script in script_content:
                match = re.search(r'max_num_pages\s*=\s*"(\d+)"', script.string)
                if match:
                    return int(match.group(1))
        except Exception:
            pass
        return 1

    def _parsear_fecha_articulo(self, texto_fecha: str) -> Optional[datetime]:
        """Convierte 'marzo 3, 2026' a datetime"""
        try:
            match = re.search(
                r'(\w+)\s+(\d{1,2}),?\s+(\d{4})', texto_fecha.lower().strip()
            )
            if match:
                mes_str, dia, año = match.group(1), int(match.group(2)), int(match.group(3))
                mes = self.MESES.get(mes_str)
                if mes:
                    return datetime(año, mes, dia)
        except Exception:
            pass
        return None

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        # Requerido por la clase base pero no se usa — la lógica está en _recolectar_links
        return []

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        links = []
        fecha_desde_dt = datetime.strptime(self.fecha_desde, '%Y-%m-%d')
        fecha_hasta_dt = datetime.strptime(self.fecha_hasta, '%Y-%m-%d')

        for pagina in range(1, total_paginas + 1):
            if pagina % 5 == 0 or pagina == 1:
                print(f"Recolectando links: página {pagina}/{total_paginas}")

            try:
                url = self._construir_url_busqueda(pagina)
                r = self.session.get(url, timeout=10)
                soup = BeautifulSoup(r.content, 'html.parser')

                contenedor = soup.find(id='tdi_89') or soup.find('div', class_=re.compile(r'tdb_loop'))
                if not contenedor:
                    continue

                modulos = contenedor.find_all(class_='td_module_wrap')

                for modulo in modulos:
                    fecha_elem = modulo.find('time')
                    fecha_art = None
                    if fecha_elem:
                        datetime_attr = fecha_elem.get('datetime', '')
                        if datetime_attr:
                            try:
                                fecha_art = datetime.fromisoformat(datetime_attr[:10])
                            except Exception:
                                pass
                        if not fecha_art:
                            fecha_art = self._parsear_fecha_articulo(fecha_elem.get_text())

                    a = modulo.find('a', rel='bookmark', href=True)
                    if not a:
                        continue

                    href = a['href']
                    if not href.startswith(self.BASE_URL):
                        continue
                    if any(x in href for x in ['/categoria/', '/author/', '/page/', '/?']):
                        continue

                    if fecha_art is None:
                        links.append(href)
                    elif fecha_art > fecha_hasta_dt:
                        continue
                    elif fecha_desde_dt <= fecha_art <= fecha_hasta_dt:
                        links.append(href)
                    elif fecha_art < fecha_desde_dt:
                        print(f"  Artículos anteriores a {self.fecha_desde}, deteniendo en página {pagina}")
                        links = list(dict.fromkeys(links))
                        print(f"Total de links únicos: {len(links)}")
                        return links

                time.sleep(0.1)

            except Exception as e:
                print(f"Error en página {pagina}: {e}")
                continue

        links = list(dict.fromkeys(links))
        print(f"Total de links únicos: {len(links)}")
        return links
```

#### **El Colombiano**

```python
class ScraperElColombiano(ScraperPeriodico):
    REGION = "eje_cafetero_antioquia"
    """Scraper específico para El Colombiano"""

    @property
    def nombre_periodico(self) -> str:
        return "El Colombiano"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        desde = datetime.strptime(self.fecha_desde, '%Y-%m-%d').strftime('%Y%m%d')
        hasta = datetime.strptime(self.fecha_hasta, '%Y-%m-%d').strftime('%Y%m%d')

        return (f"https://www.elcolombiano.com/busquedas/-/search/{self.termino}/"
               f"false/false/{desde}/{hasta}/date/true/true/0/0/meta/0/0/0/{pagina}")

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        try:
            search_header = soup.find('div', class_='searchHeader')
            if search_header:
                h2 = search_header.find('h2')
                if h2:
                    texto = h2.text.strip()
                    match = re.search(r'\((\d+)\s+resultados?\)', texto)
                    if match:
                        total_articulos = int(match.group(1))
                        return math.ceil(total_articulos / 12)
        except:
            pass

        return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        links = []
        articulos = soup.find_all('div', class_='div_iter_url')

        for articulo in articulos:
            link_elem = articulo.get('data-urldestination')
            if link_elem and link_elem != '/':
                link_completo = "https://www.elcolombiano.com" + link_elem
                links.append(link_completo)

        return links
```

```python
class ScraperBCNoticias(ScraperPeriodico):
    REGION = "eje_cafetero_antioquia"
    BASE_URL = "https://www.bcnoticias.com.co"

    def __init__(self, termino, fecha_desde, fecha_hasta):
        super().__init__(termino, fecha_desde, fecha_hasta)

    @property
    def nombre_periodico(self) -> str:
        return "BC Noticias"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        if pagina == 1:
            return f"{self.BASE_URL}/?s={self.termino}"
        return f"{self.BASE_URL}/page/{pagina}/?s={self.termino}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        try:
            pag = soup.find('div', class_='page-nav')
            if pag:
                span = pag.find('span', class_='pages')
                if span:
                    match = re.search(r'Página \d+ de (\d+)', span.get_text())
                    if match:
                        return int(match.group(1))
        except Exception:
            pass
        return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        links = []
        for modulo in soup.find_all('div', class_=lambda c: c and 'td_module_wrap' in c):
            a = modulo.find('a', rel='bookmark')
            if a and a['href'].startswith(self.BASE_URL):
                links.append(a['href'])
        return links

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        links = []
        fecha_desde_dt = datetime.strptime(self.fecha_desde, '%Y-%m-%d')
        fecha_hasta_dt = datetime.strptime(self.fecha_hasta, '%Y-%m-%d')

        for pagina in range(1, total_paginas + 1):
            if pagina % 5 == 0 or pagina == 1:
                print(f"Recolectando links: página {pagina}/{total_paginas}")

            try:
                url = self._construir_url_busqueda(pagina)
                r = self.session.get(url, timeout=10)
                soup = BeautifulSoup(r.content, 'html.parser')

                if pagina == 1:
                    total_paginas = self._obtener_total_paginas(soup)
                    print(f"Total de páginas: {total_paginas}")

                modulos = soup.find_all('div', class_=lambda c: c and 'td_module_wrap' in c)
                if not modulos:
                    print(f"  Sin resultados en página {pagina}, deteniendo")
                    break

                parar = False
                for modulo in modulos:
                    time_elem = modulo.find('time')
                    fecha_art = None
                    if time_elem:
                        try:
                            fecha_art = datetime.fromisoformat(time_elem['datetime'][:10])
                        except Exception:
                            pass

                    a = modulo.find('a', rel='bookmark')
                    if not a:
                        continue
                    href = a['href']

                    if fecha_art is None:
                        links.append(href)
                    elif fecha_art > fecha_hasta_dt:
                        continue
                    elif fecha_desde_dt <= fecha_art <= fecha_hasta_dt:
                        links.append(href)
                    elif fecha_art < fecha_desde_dt:
                        print(f"  Artículos anteriores a {self.fecha_desde}, deteniendo en página {pagina}")
                        parar = True
                        break

                if parar:
                    break

                time.sleep(0.1)

            except Exception as e:
                print(f"Error en página {pagina}: {e}")
                continue

        links = list(dict.fromkeys(links))
        print(f"Total de links únicos: {len(links)}")
        return links
```

```python
class ScraperElQuindiano(ScraperPeriodico):
    REGION = "eje_cafetero_antioquia"

    BASE_URL = "https://elquindiano.com"
    API_URL = "https://elquindiano.com/wp-json/wp/v2/posts"
    PER_PAGE = 100

    def __init__(self, termino, fecha_desde, fecha_hasta):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self._fechas = {}  # url → fecha

    @property
    def nombre_periodico(self) -> str:
        return "El Quindiano"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        return f"{self.BASE_URL}/?s={self.termino}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        return 999

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        # Requerido por la clase base pero no se usa — la lógica está en _recolectar_links
        return []

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        links = []

        # Concurrencia con asyncio y aiohttp
        async def fetch_wp_pages():
            async with aiohttp.ClientSession() as session:
                # Primera página para obtener X-WP-TotalPages
                try:
                    params = {
                        "search": self.termino, "per_page": self.PER_PAGE, "page": 1,
                        "orderby": "date", "order": "desc",
                        "after": f"{self.fecha_desde}T00:00:00", "before": f"{self.fecha_hasta}T23:59:59"
                    }
                    async with session.get(self.API_URL, params=params, timeout=15) as r:
                        if r.status != 200:
                            return []
                        total_pags = int(r.headers.get('X-WP-TotalPages', 1))
                        data = await r.json()
                        links_pagina = [post.get('link') for post in data if post.get('link')]
                        # Cache fechas
                        for post in data:
                            if post.get('link') and post.get('date'):
                                try:
                                    self._fechas[post['link']] = datetime.fromisoformat(post['date'])
                                except Exception:
                                    pass
                        links_todas = []
                        links_todas.extend(links_pagina)

                        if total_pags > 1:
                            print(f"Recolectando links de {total_pags} páginas concurrentemente...")

                            async def fetch_page(p):
                                try:
                                    p_params = params.copy()
                                    p_params["page"] = p
                                    async with session.get(self.API_URL, params=p_params, timeout=15) as rp:
                                        if rp.status == 200:
                                            p_data = await rp.json()
                                            for post in p_data:
                                                if post.get('link') and post.get('date'):
                                                    try:
                                                        self._fechas[post['link']] = datetime.fromisoformat(post['date'])
                                                    except:
                                                        pass
                                            return [post.get('link') for post in p_data if post.get('link')]
                                except Exception:
                                    pass
                                return []

                            tareas = [fetch_page(p) for p in range(2, total_pags + 1)]
                            res = await asyncio.gather(*tareas)
                            for r_pag in res:
                                links_todas.extend(r_pag)

                        return links_todas
                except Exception as e:
                    print(f"Error inicial: {e}")
                    return []

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        links = loop.run_until_complete(fetch_wp_pages())
        links = list(dict.fromkeys(links))
        print(f"Total de links únicos: {len(links)}")
        return links

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas.get(link, article.publish_date) if hasattr(self, '_fechas') else article.publish_date,
                "texto": article.text
            }
        except Exception as e:
            print(f'Error descargando: {e}')
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50) # 50 concurrent downloads
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

```python
class ScraperDiarioDelCauca(ScraperPeriodico):
    REGION = "suroccidente"
    BASE_URL = "https://diariodelcauca.com.co"
    API_URL = "https://diariodelcauca.com.co/wp-json/wp/v2/posts"
    PER_PAGE = 100

    def __init__(self, termino, fecha_desde, fecha_hasta):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self._fechas = {}
        original_get = self.session.get
        def get_con_timeout(*args, **kwargs):
            kwargs.setdefault('timeout', 30)
            return original_get(*args, **kwargs)
        self.session.get = get_con_timeout

    @property
    def nombre_periodico(self) -> str:
        return "Diario del Cauca"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        return f"{self.BASE_URL}/?s={self.termino}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        return 999

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        # Requerido por la clase base pero no se usa — la lógica está en _recolectar_links
        return []

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        links = []

        # Concurrencia con asyncio y aiohttp
        async def fetch_wp_pages():
            async with aiohttp.ClientSession() as session:
                # Primera página para obtener X-WP-TotalPages
                try:
                    params = {
                        "search": self.termino, "per_page": self.PER_PAGE, "page": 1,
                        "orderby": "date", "order": "desc",
                        "after": f"{self.fecha_desde}T00:00:00", "before": f"{self.fecha_hasta}T23:59:59"
                    }
                    async with session.get(self.API_URL, params=params, timeout=15) as r:
                        if r.status != 200:
                            return []
                        total_pags = int(r.headers.get('X-WP-TotalPages', 1))
                        data = await r.json()
                        links_pagina = [post.get('link') for post in data if post.get('link')]
                        # Cache fechas
                        for post in data:
                            if post.get('link') and post.get('date'):
                                try:
                                    self._fechas[post['link']] = datetime.fromisoformat(post['date'])
                                except Exception:
                                    pass
                        links_todas = []
                        links_todas.extend(links_pagina)

                        if total_pags > 1:
                            print(f"Recolectando links de {total_pags} páginas concurrentemente...")

                            async def fetch_page(p):
                                try:
                                    p_params = params.copy()
                                    p_params["page"] = p
                                    async with session.get(self.API_URL, params=p_params, timeout=15) as rp:
                                        if rp.status == 200:
                                            p_data = await rp.json()
                                            for post in p_data:
                                                if post.get('link') and post.get('date'):
                                                    try:
                                                        self._fechas[post['link']] = datetime.fromisoformat(post['date'])
                                                    except:
                                                        pass
                                            return [post.get('link') for post in p_data if post.get('link')]
                                except Exception:
                                    pass
                                return []

                            tareas = [fetch_page(p) for p in range(2, total_pags + 1)]
                            res = await asyncio.gather(*tareas)
                            for r_pag in res:
                                links_todas.extend(r_pag)

                        return links_todas
                except Exception as e:
                    print(f"Error inicial: {e}")
                    return []

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        links = loop.run_until_complete(fetch_wp_pages())
        links = list(dict.fromkeys(links))
        print(f"Total de links únicos: {len(links)}")
        return links

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas.get(link, article.publish_date) if hasattr(self, '_fechas') else article.publish_date,
                "texto": article.text
            }
        except Exception as e:
            print(f'Error descargando: {e}')
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50) # 50 concurrent downloads
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

#### **La República**

```python
class ScraperLaRepublica(ScraperPeriodico):
    REGION = "central"
    """Scraper específico para La República (usa Playwright Async)"""

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str, max_cargas: int = None):
        """
        Args:
            termino: Término de búsqueda
            fecha_desde: Fecha inicio (YYYY-MM-DD)
            fecha_hasta: Fecha fin (YYYY-MM-DD)
            max_cargas: Límite de veces que se hace clic en "VER MÁS" (None = sin límite)
        """
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_cargas = max_cargas

        # Permitir event loops anidados en Colab
        nest_asyncio.apply()

    @property
    def nombre_periodico(self) -> str:
        return "La República"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        """Construye URL de búsqueda con filtros de fecha"""
        # Formatear fechas para La República
        from_date = datetime.strptime(self.fecha_desde, "%Y-%m-%d").strftime("%Y-%m-%dT00:00:00-05:00")
        to_date = datetime.strptime(self.fecha_hasta, "%Y-%m-%d").strftime("%Y-%m-%dT00:00:00-05:00")

        # URL encode
        from_date = from_date.replace(":", "%3A")
        to_date = to_date.replace(":", "%3A")
        term = urllib.parse.quote(self.termino)

        return f"https://www.larepublica.co/buscar?term={term}&from={from_date}&to={to_date}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        """No aplica para La República (usa botón VER MÁS)"""
        return self.max_cargas if self.max_cargas else 999

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """Extrae enlaces de los resultados visibles"""
        enlaces = []

        try:
            resultados = soup.find_all('a', class_='result')

            for resultado in resultados:
                href = resultado.get('href', '')
                if href:
                    url = f"https://www.larepublica.co{href}" if not href.startswith('http') else href
                    enlaces.append(url)

        except Exception as e:
            print(f"Error al extraer enlaces: {e}")

        return enlaces

    async def _recolectar_links_async(self, total_paginas: int) -> List[str]:
        """Recolecta links usando Playwright Async con botón VER MÁS"""
        links = []
        cargas = 0

        async with async_playwright() as p:
            # Lanzar navegador
            browser = await p.chromium.launch(
                headless=True,
                args=['--no-sandbox', '--disable-dev-shm-usage', '--disable-gpu']
            )

            context = await browser.new_context(
                user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
            )

            page = await context.new_page()

            try:
                # Cargar página inicial
                url = self._construir_url_busqueda()

                await page.goto(url, wait_until='networkidle', timeout=30000)

                # Esperar a que carguen los resultados iniciales
                try:
                    await page.wait_for_selector('a.result', timeout=10000)
                except PlaywrightTimeout:
                    print("⚠️ No se encontraron resultados")
                    await browser.close()
                    return []

                # Extraer enlaces iniciales
                html = await page.content()
                soup = BeautifulSoup(html, 'html.parser')
                links_iniciales = self._extraer_links_pagina(soup)
                links.extend(links_iniciales)


                # Hacer clic en "VER MÁS" repetidamente
                while True:
                    # Verificar límite de cargas
                    if self.max_cargas and cargas >= self.max_cargas:

                        break

                    try:
                        # Buscar botón "VER MÁS"
                        ver_mas_button = page.locator('button.button a.btn.analisisSect:has-text("VER MÁS")')

                        # Verificar si el botón existe
                        if await ver_mas_button.count() == 0:

                            break

                        # Scroll al botón
                        await ver_mas_button.scroll_into_view_if_needed()
                        await page.wait_for_timeout(1000)

                        # Click en el botón
                        await ver_mas_button.click()
                        cargas += 1

                        # Esperar a que carguen nuevos resultados
                        await page.wait_for_timeout(3000)

                        # Extraer nuevos enlaces
                        html = await page.content()
                        soup = BeautifulSoup(html, 'html.parser')
                        enlaces_actuales = self._extraer_links_pagina(soup)

                        # Filtrar solo los nuevos
                        urls_existentes = set(links)
                        nuevos = [url for url in enlaces_actuales if url not in urls_existentes]

                        if nuevos:
                            links.extend(nuevos)

                        else:
                            break

                    except Exception as e:
                        print(f"\n✅ No se pudo hacer clic en 'VER MÁS': {e}")
                        break

            finally:
                await browser.close()

        # Eliminar duplicados
        links = list(dict.fromkeys(links))
        return links

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        """Wrapper sincrónico para el método async (override del método base)"""
        loop = asyncio.get_event_loop()
        return loop.run_until_complete(self._recolectar_links_async(total_paginas))
```

```python
class ScraperPortafolio(ScraperPeriodico):
    REGION = "central"
    """
    Scraper para Portafolio (portafolio.co).

    - HTML estático: requests + BeautifulSoup (sin Playwright)
    - Paginación por &page=N, máximo 20 artículos por página
    - Total de resultados en <p class="search__title">
    - Links relativos en <a class="page-link"> dentro de <article>
    - Fecha en <time class="c-article__date" datetime="YYYY-MM-DDTHH:MM:SS">
    - Filtro de anuncios: se excluyen links externos (syndicatedsearch, goog, etc.)
    - Parámetro max_articulos para limitar resultados (default: 300)
    """

    BASE_URL = "https://www.portafolio.co"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Referer": "https://www.portafolio.co/",
    }

    def __init__(
        self,
        termino: str,
        fecha_desde: str,
        fecha_hasta: str,
        max_articulos: int = 300,
    ):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_articulos = max_articulos
        self.session.headers.update(self.HEADERS)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Portafolio"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        termino_encoded = requests.utils.quote(self.termino)
        url = f"{self.BASE_URL}/buscar/?q={termino_encoded}"
        if pagina > 1:
            url += f"&page={pagina}"
        return url

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        try:
            p = soup.find("p", class_="search__title")
            if not p:
                return 1
            texto = p.get_text(strip=True)
            numeros = re.findall(r"\d+", texto.replace(".", "").replace(",", ""))
            if not numeros:
                return 1
            total = int(numeros[0])
            total_limitado = min(total, self.max_articulos)
            paginas = math.ceil(total_limitado / 20)
            print(f"  Total resultados: {total} | Límite: {self.max_articulos} "
                  f"| Páginas a recorrer: {paginas}")
            return paginas
        except Exception:
            return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        links = []
        articulos = soup.find_all("article", class_="c-article")

        for articulo in articulos:
            a = articulo.find("a", class_="page-link")
            if not a:
                continue

            href = a.get("href", "").strip()
            if not href:
                continue

            if href.startswith("http") and self.BASE_URL not in href:
                continue

            link_completo = self.BASE_URL + href if href.startswith("/") else href

            time_tag = articulo.find("time", class_="c-article__date")
            fecha = None
            if time_tag and time_tag.get("datetime"):
                fecha = self._parsear_fecha_portafolio(time_tag["datetime"])

            self._fechas_cache[link_completo] = fecha
            links.append(link_completo)

        return links

    # ── Parseo de fecha ────────────────────────────────────────────────────

    def _parsear_fecha_portafolio(self, datetime_str: str) -> Optional[datetime]:
        """
        Parsea el atributo datetime del tag <time>.
        Formatos esperados:
          "2023-04-15T10:30:00-05:00"
          "2023-04-15T10:30:00"
          "2023-04-15"
        Retorna datetime sin zona horaria.
        """
        if not datetime_str:
            return None

        datetime_str = datetime_str.strip()

        # Intentar con zona horaria
        for fmt in [
            "%Y-%m-%dT%H:%M:%S%z",
            "%Y-%m-%dT%H:%M:%S",
            "%Y-%m-%d",
        ]:
            try:
                dt = datetime.strptime(datetime_str[:len(fmt) + 5], fmt)
                # Quitar timezone si existe
                return dt.replace(tzinfo=None)
            except ValueError:
                continue

        # Fallback: tomar solo la parte de fecha
        try:
            return datetime.strptime(datetime_str[:10], "%Y-%m-%d")
        except ValueError:
            return None

    # ── Override de scrape() ────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  Límite de artículos: {self.max_articulos}")

        try:
            url_p1 = self._construir_url_busqueda(pagina=1)
            r = self.session.get(url_p1, timeout=15)
            r.raise_for_status()
            soup_p1 = BeautifulSoup(r.text, "html.parser")
        except Exception as e:
            print(f"  Error al cargar primera página: {e}")
            return pd.DataFrame()

        total_paginas = self._obtener_total_paginas(soup_p1)
        links_en_rango = []

        for pagina in range(1, total_paginas + 1):
            try:
                if pagina == 1:
                    soup = soup_p1
                else:
                    url = self._construir_url_busqueda(pagina=pagina)
                    r = self.session.get(url, timeout=15)
                    r.raise_for_status()
                    soup = BeautifulSoup(r.text, "html.parser")

                links_pagina = self._extraer_links_pagina(soup)
                print(f"  Página {pagina}/{total_paginas}: {len(links_pagina)} artículos")

                for link in links_pagina:
                    fecha = self._fechas_cache.get(link)
                    if fecha is None:
                        links_en_rango.append(link)
                        continue
                    if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                        links_en_rango.append(link)

                time.sleep(0.1)

            except Exception as e:
                print(f"  Error en página {pagina}: {e}")
                continue

        links_en_rango = list(dict.fromkeys(links_en_rango))
        print(f"  Total links únicos en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Descargas asíncronas ────────────────────────────────────────────────

    async def _descargar_articulo_async(
        self, session: aiohttp.ClientSession, info: tuple
    ) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                # corregido: usar _fechas_cache en vez de _fechas
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text,
            }
        except Exception as e:
            print(f"Error descargando: {e}")
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [
                self._descargar_articulo_async(session, info)
                for info in links_enumerados
            ]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(
            f"{len(data)} artículos descargados exitosamente "
            f"({len(data)/len(links)*100:.1f}%)"
        )
        return pd.DataFrame(data)
```

#### **Publimetro**

```python
class ScraperPublimetro(ScraperPeriodico):
    """
    Scraper para Publimetro Colombia (publimetro.co).

    Usa Google Custom Search Engine (CSE) — requiere Playwright para:
    - Escribir el término en el input de búsqueda
    - Esperar a que Google CSE cargue los resultados
    - Navegar por páginas haciendo clic en los números de página

    - Sin paginación por URL: navegación por botones numerados de Google CSE
    - Fecha extraída del breadcrumb: <span> › YYYY/MM/DD</span>
    - Links absolutos — filtrar /tags/ y dominios externos
    - max_articulos para limitar resultados (default: 300)
    """
    REGION = "central"
    BASE_URL     = "https://www.publimetro.co"
    SEARCH_URL   = "https://www.publimetro.co/buscador/"
    DOMINIO      = "publimetro.co"

    def __init__(
        self,
        termino: str,
        fecha_desde: str,
        fecha_hasta: str,
        max_articulos: int = 300,
    ):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_articulos = max_articulos
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Publimetro"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        return self.SEARCH_URL

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        return 1  # No aplica — manejado por Playwright

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """
        Extrae links desde <a class="gs-title"> dentro de <div class="gsc-webResult">.
        Cachea fecha desde el breadcrumb › YYYY/MM/DD.
        Filtra /tags/ y dominios externos.
        """
        links = []
        vistos = set()

        resultados = soup.find_all("div", class_="gsc-webResult")

        for resultado in resultados:
            a = resultado.find("a", class_="gs-title")
            if not a:
                continue

            href = a.get("href", "").strip()
            if not href:
                continue

            if self.DOMINIO not in href:
                continue
            if "/tags/" in href:
                continue
            if href in vistos:
                continue
            vistos.add(href)

            # Extraer fecha del breadcrumb: › YYYY/MM/DD
            fecha = None
            spans = resultado.find_all("span")
            for span in spans:
                texto = span.get_text(strip=True)
                match = re.search(r"›\s*(\d{4}/\d{2}/\d{2})", texto)
                if match:
                    fecha = self._parsear_fecha_publimetro(match.group(1))
                    break

            self._fechas_cache[href] = fecha
            links.append(href)

        return links

    # ── Parseo de fecha ────────────────────────────────────────────────────

    def _parsear_fecha_publimetro(self, fecha_str: str) -> Optional[datetime]:
        """
        Parsea fechas en formato "YYYY/MM/DD" extraídas del breadcrumb.
        Retorna datetime o None si no puede parsear.
        """
        if not fecha_str:
            return None
        try:
            return datetime.strptime(fecha_str.strip(), "%Y/%m/%d")
        except ValueError:
            return None

    # ── Playwright: búsqueda + paginación por clics ─────────────────────────

    def _get_links_con_playwright(self) -> List[str]:
        import nest_asyncio
        nest_asyncio.apply()

        async def _fetch():
            todos_links  = []
            recolectados = 0

            async with async_playwright() as p:
                browser = await p.chromium.launch(headless=True)
                page    = await browser.new_page()

                print(f"  Cargando buscador: {self.SEARCH_URL}")
                # Timeout aumentado a 60s para sitios lentos
                await page.goto(
                    self.SEARCH_URL, wait_until="networkidle", timeout=60000
                )

                input_selector = "input.gsc-input"
                await page.wait_for_selector(input_selector, timeout=15000)
                await page.fill(input_selector, self.termino)
                await page.keyboard.press("Enter")

                await page.wait_for_selector("div.gsc-webResult", timeout=20000)
                await page.wait_for_load_state("networkidle", timeout=20000)

                pagina_num = 1
                while recolectados < self.max_articulos:
                    html = await page.content()
                    soup = BeautifulSoup(html, "html.parser")

                    if pagina_num == 1:
                        info = soup.find("div", class_="gsc-result-info")
                        if info:
                            print(f"  {info.get_text(strip=True)}")

                    links_pagina = self._extraer_links_pagina(soup)
                    print(f"  Página {pagina_num}: {len(links_pagina)} artículos")
                    todos_links.extend(links_pagina)
                    recolectados += len(links_pagina)

                    if recolectados >= self.max_articulos:
                        print(f"  Límite de {self.max_articulos} artículos alcanzado.")
                        break

                    siguiente_pagina = pagina_num + 1
                    btn_selector = (
                        f"div.gsc-cursor-page[aria-label='Page {siguiente_pagina}']"
                    )
                    btn   = page.locator(btn_selector)
                    count = await btn.count()

                    if count == 0:
                        print("  Sin más páginas disponibles.")
                        break

                    await btn.first.click()
                    await page.wait_for_load_state("networkidle", timeout=20000)
                    await page.wait_for_selector(
                        "div.gsc-webResult", timeout=15000
                    )
                    pagina_num += 1

                await browser.close()
            return todos_links

        loop = asyncio.get_event_loop()
        return loop.run_until_complete(_fetch())

    # ── Override de scrape() ────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  Límite de artículos: {self.max_articulos}")

        try:
            todos_links = self._get_links_con_playwright()
        except Exception as e:
            print(f"  Error en Playwright: {e}")
            return pd.DataFrame()

        links_unicos = list(dict.fromkeys(todos_links))
        print(f"  Links únicos recolectados: {len(links_unicos)}")

        # Filtrar por rango de fechas
        links_en_rango = []
        for link in links_unicos:
            fecha = self._fechas_cache.get(link)
            if fecha is None:
                links_en_rango.append(link)
                continue
            if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                links_en_rango.append(link)

        print(f"  Links en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Descargas asíncronas ────────────────────────────────────────────────

    async def _descargar_articulo_async(
        self, session: aiohttp.ClientSession, info: tuple
    ) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                # corregido: usar _fechas_cache en vez de _fechas
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text,
            }
        except Exception as e:
            print(f"Error descargando: {e}")
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [
                self._descargar_articulo_async(session, info)
                for info in links_enumerados
            ]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(
            f"{len(data)} artículos descargados exitosamente "
            f"({len(data)/len(links)*100:.1f}%)"
        )
        return pd.DataFrame(data)
```

#### **Las2Orillas**

```python
class ScraperLas2Orillas(ScraperPeriodico):
    """
    Scraper para Las2Orillas (las2orillas.co).

    - HTML estático: requests + BeautifulSoup (sin Playwright)
    - Paginación por URL: https://www.las2orillas.co/page/N/?s=termino
    - Total de páginas extraído del último <a class="page-numbers">
    - 8 artículos por página
    - Links absolutos en <h4 class="entry-title title"> > <a>
    - Fecha en <a href="...las2orillas.co/YYYY/MM/...">mes DD, YYYY</a>
    - Filtro: solo links dentro de las2orillas.co sin /contacto-comercial/,
      /wp-content/, /c/, /author/, /page/, /category/
    - max_articulos para limitar resultados (default: 300)
    """
    REGION = "central"
    BASE_URL   = "https://www.las2orillas.co"
    DOMINIO    = "las2orillas.co"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Referer": "https://www.las2orillas.co/",
    }

    PATHS_EXCLUIDOS = (
        "/contacto-comercial/",
        "/wp-content/",
        "/author/",
        "/category/",
        "/c/",
        "/page/",
        "/tag/",
    )

    MESES_ES = {
        "enero": 1, "febrero": 2, "marzo": 3, "abril": 4,
        "mayo": 5, "junio": 6, "julio": 7, "agosto": 8,
        "septiembre": 9, "octubre": 10, "noviembre": 11, "diciembre": 12,
    }

    def __init__(
        self,
        termino: str,
        fecha_desde: str,
        fecha_hasta: str,
        max_articulos: int = 300,
    ):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_articulos = max_articulos
        self.session.headers.update(self.HEADERS)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Las2Orillas"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        termino_encoded = requests.utils.quote(self.termino).replace("%20", "+")
        if pagina == 1:
            return f"{self.BASE_URL}/?s={termino_encoded}"
        return f"{self.BASE_URL}/page/{pagina}/?s={termino_encoded}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        try:
            paginas = soup.find_all("a", class_="page-numbers")
            numeros = []
            for a in paginas:
                texto = a.get_text(strip=True)
                if texto.isdigit():
                    numeros.append(int(texto))

            if not numeros:
                return 1

            total_paginas_sitio  = max(numeros)
            max_paginas_por_limite = math.ceil(self.max_articulos / 8)
            total_paginas = min(total_paginas_sitio, max_paginas_por_limite)

            print(f"  Total páginas en sitio: {total_paginas_sitio} | "
                  f"Límite: {self.max_articulos} arts → recorriendo {total_paginas} páginas")
            return total_paginas
        except Exception:
            return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        links = []
        articulos = soup.find_all("article", class_="mg-posts-sec-post")

        for articulo in articulos:
            h4 = articulo.find("h4", class_="entry-title")
            if not h4:
                continue

            a = h4.find("a")
            if not a:
                continue

            href = a.get("href", "").strip()
            if not href:
                continue

            if self.DOMINIO not in href:
                continue
            if any(path in href for path in self.PATHS_EXCLUIDOS):
                continue

            fecha = None
            span_fecha = articulo.find("span", class_="mg-blog-date")
            if span_fecha:
                a_fecha = span_fecha.find("a")
                if a_fecha:
                    texto_fecha = a_fecha.get_text(strip=True)
                    fecha = self._parsear_fecha_las2orillas(texto_fecha)

            self._fechas_cache[href] = fecha
            links.append(href)

        return links

    # ── Parseo de fecha ────────────────────────────────────────────────────

    def _parsear_fecha_las2orillas(self, texto: str) -> Optional[datetime]:
        """
        Parsea fechas en formato español: "agosto 13, 2024" o "agosto 13 2024".
        Retorna datetime o None si no puede parsear.
        """
        if not texto:
            return None

        texto = texto.lower().strip().replace(",", "")
        partes = texto.split()

        if len(partes) < 3:
            return None

        try:
            mes_str = partes[0]
            dia     = int(partes[1])
            anio    = int(partes[2])
            mes     = self.MESES_ES.get(mes_str)

            if mes is None:
                return None

            return datetime(anio, mes, dia)
        except (ValueError, IndexError):
            return None

    # ── Override de scrape() ────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  Límite de artículos: {self.max_articulos}")

        try:
            url_p1 = self._construir_url_busqueda(pagina=1)
            r = self.session.get(url_p1, timeout=30)  # timeout aumentado
            r.raise_for_status()
            soup_p1 = BeautifulSoup(r.text, "html.parser")
        except Exception as e:
            print(f"  Error al cargar primera página: {e}")
            return pd.DataFrame()

        total_paginas = self._obtener_total_paginas(soup_p1)
        links_en_rango = []
        errores_consecutivos = 0

        for pagina in range(1, total_paginas + 1):
            try:
                if pagina == 1:
                    soup = soup_p1
                else:
                    url = self._construir_url_busqueda(pagina=pagina)

                    respuesta = None
                    for intento in range(3):
                        try:
                            r = self.session.get(url, timeout=30)
                            r.raise_for_status()
                            respuesta = r
                            break
                        except Exception:
                            espera = 5 * (intento + 1)
                            print(f"  Intento {intento+1}/3 fallido en página {pagina} "
                                  f"— esperando {espera}s...")
                            time.sleep(espera)

                    if respuesta is None:
                        errores_consecutivos += 1
                        print(f"  Página {pagina} descartada tras 3 intentos.")
                        if errores_consecutivos >= 5:
                            print("  5 páginas consecutivas fallidas — deteniendo.")
                            break
                        continue

                    errores_consecutivos = 0
                    soup = BeautifulSoup(respuesta.text, "html.parser")

                links_pagina = self._extraer_links_pagina(soup)
                print(f"  Página {pagina}/{total_paginas}: {len(links_pagina)} artículos")

                for link in links_pagina:
                    fecha = self._fechas_cache.get(link)
                    if fecha is None:
                        links_en_rango.append(link)
                        continue
                    if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                        links_en_rango.append(link)

                time.sleep(1.5)

            except Exception as e:
                print(f"  Error en página {pagina}: {e}")
                errores_consecutivos += 1
                if errores_consecutivos >= 5:
                    print("  5 páginas consecutivas fallidas — deteniendo.")
                    break
                continue

        links_en_rango = list(dict.fromkeys(links_en_rango))
        print(f"  Total links únicos en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Descargas asíncronas ────────────────────────────────────────────────

    async def _descargar_articulo_async(
        self, session: aiohttp.ClientSession, info: tuple
    ) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                # corregido: usar _fechas_cache en vez de _fechas
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text,
            }
        except Exception as e:
            print(f"Error descargando: {e}")
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [
                self._descargar_articulo_async(session, info)
                for info in links_enumerados
            ]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(
            f"{len(data)} artículos descargados exitosamente "
            f"({len(data)/len(links)*100:.1f}%)"
        )
        return pd.DataFrame(data)
```

```python
class ScraperElHeraldo(ScraperPeriodico):
    """Scraper específico para El Heraldo (usa Playwright Async)"""
    REGION = "caribe"

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str, max_paginas: int = None):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_paginas = max_paginas
        self.inicio_rango = datetime.strptime(fecha_desde, '%Y-%m-%d')
        self.fin_rango = datetime.strptime(fecha_hasta, '%Y-%m-%d') + timedelta(days=1) - timedelta(seconds=1)

        # Permitir event loops anidados en Colab
        nest_asyncio.apply()

    @property
    def nombre_periodico(self) -> str:
        return "El Heraldo"

    def _parse_spanish_date(self, date_str: str) -> Optional[datetime]:
        """Convierte fecha en formato español a datetime"""
        meses_es = {
            'ene': '01', 'feb': '02', 'mar': '03', 'abr': '04', 'may': '05', 'jun': '06',
            'jul': '07', 'ago': '08', 'sep': '09', 'oct': '10', 'nov': '11', 'dic': '12'
        }

        match = re.search(r'(\d{1,2})\s+(\w{3})\s+(\d{4})', date_str.lower())

        if match:
            dia, mes_abbr, anio = match.groups()
            mes_num = meses_es.get(mes_abbr)

            if mes_num:
                try:
                    return datetime.strptime(f"{dia.zfill(2)}/{mes_num}/{anio}", '%d/%m/%Y')
                except ValueError:
                    return None
        return None

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        """Construye URL de búsqueda"""
        keywords_url = self.termino.replace(" ", "%20").replace(",", "%2C")
        return f"https://www.elheraldo.co/buscador/?query={keywords_url}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        """No aplica para El Heraldo (usa carga dinámica)"""
        return self.max_paginas if self.max_paginas else 999

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """Extrae enlaces con filtrado por fecha"""
        enlaces = []

        try:
            items = soup.find_all('div', class_='queryly_item_row')

            for item in items:
                link_element = item.find('a')
                if link_element and link_element.get('href'):
                    href = link_element['href']
                    url_completa = f"https://www.elheraldo.co{href}" if not href.startswith('http') else href

                    # Extraer y filtrar por fecha
                    fecha_div = item.find('div', style=lambda value: value and 'color:#555' in value)
                    fecha_texto = fecha_div.get_text(strip=True) if fecha_div else ""

                    parsed_date = self._parse_spanish_date(fecha_texto)

                    if parsed_date and self.inicio_rango <= parsed_date <= self.fin_rango:
                        enlaces.append(url_completa)

        except Exception as e:
            print(f"Error al extraer enlaces: {e}")

        return enlaces

    async def _recolectar_links_async(self, total_paginas: int) -> List[str]:
        """Recolecta links usando Playwright Async"""
        links = []
        pagina = 1

        async with async_playwright() as p:
            # Lanzar navegador
            browser = await p.chromium.launch(
                headless=True,
                args=['--no-sandbox', '--disable-dev-shm-usage', '--disable-gpu']
            )

            context = await browser.new_context(
                user_agent='Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
            )

            page = await context.new_page()

            try:
                # Cargar página inicial
                url = self._construir_url_busqueda()

                await page.goto(url, wait_until='networkidle', timeout=30000)

                while True:

                    # Esperar a que carguen los resultados
                    try:
                        await page.wait_for_selector('.queryly_item_row', timeout=10000)
                    except PlaywrightTimeout:

                        break

                    # Extraer HTML y parsear
                    html = await page.content()
                    soup = BeautifulSoup(html, 'html.parser')
                    links_pagina = self._extraer_links_pagina(soup)

                    if links_pagina:
                        links.extend(links_pagina)

                    else:
                        None

                    # Verificar límite de páginas
                    if self.max_paginas and pagina >= self.max_paginas:
                        break

                    # Buscar botón "Siguiente"
                    try:
                        next_button = page.locator('a.next_btn')

                        # Verificar si el botón existe
                        if await next_button.count() == 0:
                            break

                        # Scroll al botón
                        await next_button.scroll_into_view_if_needed()
                        await page.wait_for_timeout(500)

                        # Click en el botón
                        await next_button.click()

                        # Esperar a que cargue nuevo contenido
                        await page.wait_for_timeout(2000)

                        pagina += 1

                    except Exception as e:
                        print(f"\n✅ No se pudo hacer clic en 'Siguiente': {e}")
                        break

            finally:
                await browser.close()

        # Eliminar duplicados
        links = list(dict.fromkeys(links))
        print(f"Total de links únicos: {len(links)}")
        return links

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        """Wrapper sincrónico para el método async (override del método base)"""
        loop = asyncio.get_event_loop()
        return loop.run_until_complete(self._recolectar_links_async(total_paginas))
```

#### **El Universal**

```python
class ScraperElUniversal(ScraperPeriodico):
    """
    Scraper para El Universal (eluniversal.com.co).

    Usa Queryly API con sort=date para orden cronológico descendente.
    Búsqueda binaria para saltar directamente a la página del rango,
    evitando recorrer cientos de páginas de artículos más recientes.
    """

    BASE_URL = "https://www.eluniversal.com.co"
    QUERYLY_KEY = "83abeafa666b4fc1"
    BATCH_SIZE = 20
    REGION = "caribe"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Referer": "https://www.eluniversal.com.co/",
        "Accept": "application/json, text/javascript, */*; q=0.01",
        "Accept-Language": "es-CO,es;q=0.9,en;q=0.8",
    }

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.session.headers.update(self.HEADERS)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._total_resultados: Optional[int] = None

    @property
    def nombre_periodico(self) -> str:
        return "El Universal"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        endindex = (pagina - 1) * self.BATCH_SIZE
        params = {
            "queryly_key": self.QUERYLY_KEY,
            "initialized": "1",
            "query": self.termino,
            "endindex": endindex,
            "batchsize": self.BATCH_SIZE,
            "callback": "",
            "extendeddatafields": "creator,imageresizer,promo_image",
            "timezoneoffset": "300",
            "sort": "date",
        }
        req = requests.Request("GET", "https://api.queryly.com/v4/search.aspx", params=params)
        return req.prepare().url

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        if self._total_resultados is None:
            return 1
        return math.ceil(self._total_resultados / self.BATCH_SIZE)

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        return getattr(self, "_links_pagina_actual", [])

    # ── Búsqueda binaria ────────────────────────────────────────────────────

    def _encontrar_pagina_inicio(self, total_paginas: int) -> int:
        """
        Búsqueda binaria para encontrar la primera página que toca
        el rango de fechas. Encuentra el punto en ~log2(total_paginas)
        requests en lugar de recorrer secuencialmente desde la página 1.
        """
        izq, der = 1, total_paginas
        resultado = total_paginas

        print(f"  Buscando página de inicio (búsqueda binaria sobre {total_paginas} páginas)...")

        while izq <= der:
            mid = (izq + der) // 2
            try:
                url = self._construir_url_busqueda(mid)
                r = self.session.get(url, timeout=15)
                items, _ = self._parsear_respuesta_queryly(r.text)
                fechas = [self._parsear_fecha_queryly(i.get("pubdate", "")) for i in items]
                fechas_validas = [f for f in fechas if f]

                if not fechas_validas:
                    der = mid - 1
                    continue

                f_min = min(fechas_validas)
                f_max = max(fechas_validas)

                print(f"    Página {mid:>4}: {f_min.strftime('%Y-%m-%d')} → {f_max.strftime('%Y-%m-%d')}")

                if f_max < self._fecha_desde_dt:
                    # Página completamente anterior al rango — buscar más atrás
                    resultado = mid
                    der = mid - 1
                elif f_min > self._fecha_hasta_dt:
                    # Página completamente posterior al rango — avanzar
                    izq = mid + 1
                else:
                    # Página toca el rango — puede haber inicio más atrás
                    resultado = mid
                    der = mid - 1

                time.sleep(0.2)

            except Exception as e:
                print(f"  [búsqueda binaria] Error en página {mid}: {e}")
                izq = mid + 1

        # Margen de seguridad de 2 páginas para no perder artículos del borde
        pagina_inicio = max(1, resultado - 2)
        print(f"  Página de inicio encontrada: {pagina_inicio}")
        return pagina_inicio

    # ── Recolección de links ────────────────────────────────────────────────

    def _recolectar_links(self, total_paginas: int) -> List[str]:
        links = []
        pagina_inicio = self._encontrar_pagina_inicio(total_paginas)
        paginas_consecutivas_viejas = 0

        for pagina in range(pagina_inicio, total_paginas + 1):
            if pagina % 10 == 0 or pagina == pagina_inicio:
                print(f"  Recolectando links: página {pagina}/{total_paginas}")

            try:
                url = self._construir_url_busqueda(pagina)
                r = self.session.get(url, timeout=15)
                items, _ = self._parsear_respuesta_queryly(r.text)

                if not items:
                    print(f"  Página {pagina}: sin resultados, deteniendo.")
                    break

                links_pagina, _ = self._filtrar_links_por_fecha(items)
                links.extend(links_pagina)

                fechas = [self._parsear_fecha_queryly(i.get("pubdate", "")) for i in items]
                fechas_validas = [f for f in fechas if f]

                if fechas_validas:
                    if max(fechas_validas) < self._fecha_desde_dt:
                        paginas_consecutivas_viejas += 1
                        if paginas_consecutivas_viejas >= 2:
                            print(f"  Artículos anteriores al rango. Deteniendo en página {pagina}.")
                            break
                    else:
                        paginas_consecutivas_viejas = 0

                time.sleep(0.3)

            except Exception as e:
                print(f"  Error en página {pagina}: {e}")
                continue

        links = list(dict.fromkeys(links))
        print(f"Total de links únicos en rango de fechas: {len(links)}")
        return links

    # ── scrape() ────────────────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")

        try:
            url_inicial = self._construir_url_busqueda(pagina=1)
            r = self.session.get(url_inicial, timeout=15)
            _, total = self._parsear_respuesta_queryly(r.text)
            total_paginas = math.ceil(total / self.BATCH_SIZE)
            print(f"  Total artículos en Queryly: {total:,} | Páginas estimadas: {total_paginas}")
        except Exception as e:
            print(f"  Error obteniendo total de resultados: {e}")
            return pd.DataFrame()

        links = self._recolectar_links(total_paginas)

        if not links:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links)

    # ── Métodos auxiliares ──────────────────────────────────────────────────

    def _parsear_respuesta_queryly(self, texto: str):
        match = re.search(r"JSON\.parse\('(.+?)'\);", texto, re.DOTALL)
        if not match:
            raise ValueError("No se encontró JSON en la respuesta de Queryly")
        raw_json = match.group(1).encode("utf-8").decode("unicode_escape")
        data = json.loads(raw_json)
        items = data.get("items", [])
        total = data.get("metadata", {}).get("total", 0)
        if self._total_resultados is None:
            self._total_resultados = total
        return items, total

    def _filtrar_links_por_fecha(self, items: list):
        links_en_rango = []
        fuera_de_rango = 0
        for item in items:
            link_relativo = item.get("link", "")
            if not self._es_link_articulo(link_relativo):
                continue
            fecha_articulo = self._parsear_fecha_queryly(item.get("pubdate", ""))
            if fecha_articulo is None:
                links_en_rango.append(self.BASE_URL + link_relativo)
                continue
            if self._fecha_desde_dt <= fecha_articulo <= self._fecha_hasta_dt:
                links_en_rango.append(self.BASE_URL + link_relativo)
            else:
                fuera_de_rango += 1
        return links_en_rango, fuera_de_rango

    def _es_link_articulo(self, link: str) -> bool:
        if not link or not link.startswith("/") or link.startswith("http"):
            return False
        segmentos = [s for s in link.strip("/").split("/") if s]
        if len(segmentos) < 4:
            return False
        try:
            año = int(segmentos[1])
            return 2000 <= año <= 2030
        except (ValueError, IndexError):
            return False

    @staticmethod
    def _parsear_fecha_queryly(pubdate: str) -> Optional[datetime]:
        if not pubdate:
            return None
        meses_es = {
            "ene": "Jan", "feb": "Feb", "mar": "Mar", "abr": "Apr",
            "may": "May", "jun": "Jun", "jul": "Jul", "ago": "Aug",
            "sep": "Sep", "oct": "Oct", "nov": "Nov", "dic": "Dec"
        }
        pubdate_norm = pubdate.strip()
        for es, en in meses_es.items():
            pubdate_norm = re.sub(rf"\b{es}\b", en, pubdate_norm, flags=re.IGNORECASE)
        for fmt in ["%b %d, %Y", "%B %d, %Y"]:
            try:
                return datetime.strptime(pubdate_norm, fmt)
            except ValueError:
                continue
        return None
```

#### **El Pilón**

```python
class ScraperElPilon(ScraperPeriodico):
    """
    Scraper para El Pilón (elpilon.com.co).

    - Búsqueda: ?q=termino (una sola página, máximo 15 artículos)
    - Sin paginación
    - HTML dinámico — Playwright para renderizar Next.js
    - Fecha extraída del <time datetime="..."> en la página de resultados
    - Links relativos → se completan con BASE_URL
    - Sin filtro de fechas nativo — postprocesamiento
    """

    BASE_URL = "https://elpilon.com.co"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Accept-Language": "es-CO,es;q=0.9,en;q=0.8",
        "Referer": "https://elpilon.com.co/",
    }
    REGION = "caribe"

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.session.headers.update(self.HEADERS)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "El Pilón"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        """
        El Pilón solo tiene una página de resultados.
        El parámetro 'pagina' se ignora — siempre retorna la misma URL.
        """
        termino_encoded = requests.utils.quote(self.termino, safe="")
        return f"{self.BASE_URL}/buscar?q={termino_encoded}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        """El Pilón no tiene paginación — siempre es 1 página."""
        return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        links = []
        vistos = set()

        for time_tag in soup.find_all("time", datetime=re.compile(r"^\d{4}-\d{2}-\d{2}T")):
            contenedor = time_tag.find_parent("div")
            if not contenedor:
                continue

            a_titulo = None
            for a in contenedor.find_all("a", href=True):
                href = a.get("href", "")
                segmentos = [s for s in href.strip("/").split("/") if s]
                if href.startswith("/") and len(segmentos) >= 2:
                    a_titulo = a
                    break

            if not a_titulo:
                continue

            href = a_titulo.get("href").strip()
            link_completo = self.BASE_URL + href

            if link_completo in vistos:
                continue
            vistos.add(link_completo)

            fecha = self._parsear_fecha_iso(time_tag.get("datetime", ""))
            self._fechas_cache[link_completo] = fecha
            links.append(link_completo)

        return links

    # ── Playwright: wait_until corregido a domcontentloaded ─────────────────

    def _get_html_con_playwright(self, url: str) -> str:
        async def _fetch(url):
            async with async_playwright() as p:
                browser = await p.chromium.launch(headless=True)
                page = await browser.new_page()
                # FIX: "networkidle" causa timeout en Next.js — usar "domcontentloaded"
                await page.goto(url, wait_until="domcontentloaded", timeout=30000)
                try:
                    await page.wait_for_selector("time[datetime]", timeout=15000)
                except Exception:
                    pass
                html = await page.content()
                await browser.close()
                return html

        return asyncio.get_event_loop().run_until_complete(_fetch(url))

    # ── scrape(): código duplicado eliminado ────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  [INFO] Máximo 15 artículos por búsqueda (limitación del sitio)")
        print(f"  [INFO] Usando Playwright — sitio Next.js (contenido dinámico)")

        try:
            url = self._construir_url_busqueda()
            html = self._get_html_con_playwright(url)
            soup = BeautifulSoup(html, "html.parser")
            links_todos = self._extraer_links_pagina(soup)
            print(f"  Links encontrados: {len(links_todos)}")
        except Exception as e:
            print(f"  Error obteniendo artículos: {e}")
            return pd.DataFrame()

        links_en_rango = []
        for link in links_todos:
            fecha = self._fechas_cache.get(link)
            if fecha is None:
                links_en_rango.append(link)
                continue
            if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                links_en_rango.append(link)

        print(f"  Links en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Descargas asíncronas ─────────────────────────────────────────────────

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text
            }
        except Exception as e:
            print(f"  Error descargando {link}: {e}")
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))
        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

#### **El Meridiano**

```python
class ScraperElMeridiano(ScraperPeriodico):
    """
    Scraper para El Meridiano (elmeridiano.co).

    El sitio cubre dos regiones: Córdoba y Sucre. Se recorren ambas
    búsquedas y se deduplicam los links resultantes.

    - Sitio Next.js — requiere Playwright para renderizar contenido
    - Sin paginación por URL: el botón "Siguiente" es JavaScript puro
      → se hace clic programáticamente con Playwright
    - Máximo 5 artículos por página
    - Fecha extraída del <p> con formato "Sección • DD/MM/YYYY"
    - Links relativos → se completan con BASE_URL
    """

    BASE_URL = "https://elmeridiano.co"
    REGIONES = ["cordoba", "sucre"]
    REGION = "caribe"

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "El Meridiano"

    def _construir_url_busqueda(self, pagina: int = 1, region: str = "cordoba") -> str:
        termino_encoded = requests.utils.quote(self.termino)
        return f"{self.BASE_URL}/busqueda?q={termino_encoded}&site={region}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """
        Extrae links desde <a class="block p-4 border...">.
        Si el selector original falla (0 resultados), hace fallback
        buscando cualquier <a> con href de artículo válido.
        """
        links = []

        # Intento 1: selector original con clases Tailwind
        articulos = soup.find_all(
            "a",
            class_=lambda c: c and "block" in c and "border" in c and "rounded-lg" in c
        )

        # Intento 2: fallback — cualquier <a> con href que parezca artículo
        if not articulos:
            articulos = [
                a for a in soup.find_all("a", href=True)
                if self._es_link_articulo_meridiano(a.get("href", ""))
            ]

        for a in articulos:
            href = a.get("href", "").strip()
            if not href or not href.startswith("/"):
                continue

            link_completo = self.BASE_URL + href

            fecha = None
            for p in a.find_all("p"):
                texto = p.get_text(strip=True)
                if "•" in texto:
                    fecha = self._parsear_fecha_meridiano(texto)
                    break

            self._fechas_cache[link_completo] = fecha
            links.append(link_completo)

        return links

    def _es_link_articulo_meridiano(self, href: str) -> bool:
        """Valida que el href parezca un artículo y no una sección."""
        if not href or not href.startswith("/"):
            return False
        segmentos = [s for s in href.strip("/").split("/") if s]
        return len(segmentos) >= 2

    # ── Playwright: wait_until corregido a domcontentloaded ─────────────────

    def _get_links_con_playwright(self, region: str) -> List[str]:
        async def _fetch():
            links_region = []

            async with async_playwright() as p:
                browser = await p.chromium.launch(headless=True)
                page = await browser.new_page()

                url = self._construir_url_busqueda(region=region)
                print(f"    [{region.upper()}] Cargando: {url}")

                # FIX: "networkidle" causa timeout en Next.js — usar "domcontentloaded"
                await page.goto(url, wait_until="domcontentloaded", timeout=30000)

                pagina_num = 1
                while True:
                    # Esperar artículos con timeout generoso
                    try:
                        await page.wait_for_selector("a.block", timeout=10000)
                    except Exception:
                        print(f"    [{region.upper()}] Página {pagina_num}: sin artículos (timeout selector)")
                        break

                    html = await page.content()
                    soup = BeautifulSoup(html, "html.parser")
                    links_pagina = self._extraer_links_pagina(soup)
                    print(f"    [{region.upper()}] Página {pagina_num}: {len(links_pagina)} artículos")
                    links_region.extend(links_pagina)

                    # Buscar botón "Siguiente" habilitado
                    btn_siguiente = page.locator("button:has-text('Siguiente'):not([disabled])")
                    count = await btn_siguiente.count()

                    if count == 0:
                        print(f"    [{region.upper()}] Sin más páginas")
                        break

                    await btn_siguiente.first.click()
                    # FIX: después del clic también usar domcontentloaded
                    await page.wait_for_load_state("networkidle", timeout=15000)
                    pagina_num += 1

                await browser.close()
            return links_region

        return asyncio.get_event_loop().run_until_complete(_fetch())

    # ── Override de scrape() ────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  [INFO] Recorriendo ambas regiones: Córdoba y Sucre")

        todos_los_links = []
        for region in self.REGIONES:
            try:
                links_region = self._get_links_con_playwright(region)
                todos_los_links.extend(links_region)
            except Exception as e:
                print(f"  Error en región {region}: {e}")
                continue

        links_unicos = list(dict.fromkeys(todos_los_links))
        print(f"\n  Links totales (sin duplicados de URL): {len(links_unicos)}")

        links_en_rango = []
        for link in links_unicos:
            fecha = self._fechas_cache.get(link)
            if fecha is None:
                links_en_rango.append(link)
                continue
            if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                links_en_rango.append(link)

        print(f"  Links en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        df = self._descargar_articulos(links_en_rango)

        if not df.empty and "titulo" in df.columns:
            antes = len(df)
            df = df.drop_duplicates(subset=["titulo"], keep="first")
            eliminados = antes - len(df)
            if eliminados > 0:
                print(f"  Duplicados por título eliminados: {eliminados}")

        return df

    # ── Descargas asíncronas ─────────────────────────────────────────────────

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text
            }
        except Exception as e:
            print(f"  Error descargando {link}: {e}")
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))
        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

#### **Vanguardia**

```python
class ScraperVanguardia(ScraperPeriodico):
    """
    Scraper para Vanguardia (vanguardia.com).

    Usa Queryly como motor de búsqueda externo (igual que El Universal).
    La API retorna JSONP — se extrae el JSON embebido con regex.

    - API: api.queryly.com con queryly_key=71d6147f75ed47b0
    - Paginación: endindex avanza de 20 en 20
    - Links relativos → se completan con BASE_URL
    - Fecha en campo 'pubdate': formato "Apr 22, 2024"
    - Filtro de fechas por postprocesamiento (Queryly no filtra por fecha)
    """

    BASE_URL        = "https://www.vanguardia.com"
    QUERYLY_KEY     = "71d6147f75ed47b0"
    BATCH_SIZE      = 20
    API_URL         = "https://api.queryly.com/json.aspx"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Referer": "https://www.vanguardia.com/",
    }

    REGION = "nororiente"

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.session.headers.update(self.HEADERS)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Vanguardia"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        """No se usa directamente — la paginación se maneja por endindex en la API."""
        return self.BASE_URL

    def _construir_url_api(self, endindex: int) -> str:
        termino_encoded = requests.utils.quote(self.termino)
        params = (
            f"queryly_key={self.QUERYLY_KEY}"
            f"&query={termino_encoded}"
            f"&endindex={endindex}"
            f"&batchsize={self.BATCH_SIZE}"
            f"&callback=searchPage.resultcallback"
            f"&showfaceted=true"
            f"&extendeddatafields=creator,imageresizer,promo_image"
            f"&timezoneoffset=300"
        )
        return f"{self.API_URL}?{params}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        # No aplica — paginación manejada directamente en scrape()
        return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        # No aplica — extracción desde JSON de la API
        return []

    # ── Lógica principal: consumir API Queryly ──────────────────────────────

    def _consultar_api(self, endindex: int) -> dict:
      url = self._construir_url_api(endindex)
      r = self.session.get(url, timeout=15)
      r.raise_for_status()

      # Respuesta JSONP: try{ searchPage.resultcallback({...}); } catch(e){}
      # raw_decode parsea el JSON y descarta el ')' sobrante al final
      inicio = r.text.find("(")
      if inicio == -1:
          raise ValueError(f"Respuesta inesperada de Queryly: {r.text[:200]}")

      decoder = json.JSONDecoder()
      data, _ = decoder.raw_decode(r.text, inicio + 1)
      return data

    def _procesar_items(self, items: list) -> List[str]:
        """
        Extrae links del batch de items y cachea fechas.
        Retorna lista de URLs absolutas.
        """
        links = []
        for item in items:
            href = item.get("link", "").strip()
            if not href:
                continue

            # Completar link relativo
            link_completo = self.BASE_URL + href if href.startswith("/") else href

            # Parsear y cachear fecha
            fecha = self._parsear_fecha_queryly(item.get("pubdate", ""))
            self._fechas_cache[link_completo] = fecha
            links.append(link_completo)

        return links

    def _parsear_fecha_queryly(self, pubdate: str) -> Optional[datetime]:
        """
        Parsea la fecha del campo 'pubdate' de la API Queryly.
        Formato esperado: "Apr 22, 2024"
        """
        if not pubdate:
            return None
        try:
            return datetime.strptime(pubdate.strip(), "%b %d, %Y")
        except ValueError:
            pass
        # Intentar formato alternativo con hora: "Apr 22, 2024 10:30 AM"
        try:
            return datetime.strptime(pubdate.strip()[:12], "%b %d, %Y")
        except ValueError:
            pass
        return None

    # ── Override de scrape() ────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")

        # Primera llamada para obtener total de resultados
        try:
            data = self._consultar_api(endindex=0)
            total = int(data.get("metadata", {}).get("total", 0))
            print(f"  Total de resultados en Queryly: {total}")
        except Exception as e:
            print(f"  Error en primera consulta a la API: {e}")
            return pd.DataFrame()

        if total == 0:
            print("  No se encontraron resultados.")
            return pd.DataFrame()

        # Recolectar links paginando
        links_en_rango = []
        detener = False

        MAX_RESULTADOS = 300

        for endindex in range(0, min(total, MAX_RESULTADOS), self.BATCH_SIZE):
            if detener:
                break

            try:
                if endindex > 0:
                    data = self._consultar_api(endindex)

                items = data.get("items", [])
                if not items:
                    break

                links_batch = self._procesar_items(items)
                pagina_actual = endindex // self.BATCH_SIZE + 1
                print(f"  Página {pagina_actual}: {len(links_batch)} artículos")

                for link in links_batch:
                    fecha = self._fechas_cache.get(link)

                    if fecha is None:
                        links_en_rango.append(link)
                        continue

                    if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                        links_en_rango.append(link)

                time.sleep(0.3)

            except Exception as e:
                print(f"  Error en endindex={endindex}: {e}")
                continue

        links_en_rango = list(dict.fromkeys(links_en_rango))
        print(f"  Total links únicos en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Override de _descargar_articulo_individual (newspaper4k) ────────────

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas.get(link, article.publish_date) if hasattr(self, '_fechas') else article.publish_date,
                "texto": article.text
            }
        except Exception as e:
            print(f'Error descargando: {e}')
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50) # 50 concurrent downloads
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

#### **Trochando Sin Fronteras**

```python
class ScraperTrochandoSinFronteras(ScraperPeriodico):
    """
    Scraper para Trochando Sin Fronteras (trochandosinfronteras.info).

    - HTML estático: requests + BeautifulSoup (sin Playwright)
    - Paginación por URL: /page/N/?s=termino
    - Total de páginas desde <a class="last" title="N">N</a>
    - 10 artículos por página
    - Links en <h3 class="entry-title td-module-title"> > <a>
    - Solo dentro de <div class="tdi_96 ..."> (columna principal, no sidebar)
    - Fecha en <time class="entry-date updated td-module-date" datetime="...">
    - max_articulos para limitar resultados (default: 300)
    """

    BASE_URL = "https://trochandosinfronteras.info"
    DOMINIO  = "trochandosinfronteras.info"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Referer": "https://trochandosinfronteras.info/",
    }

    REGION = "nororiente"

    def __init__(
        self,
        termino: str,
        fecha_desde: str,
        fecha_hasta: str,
        max_articulos: int = 300,
    ):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_articulos = max_articulos
        self.session.headers.update(self.HEADERS)
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Trochando Sin Fronteras"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        termino_encoded = requests.utils.quote(self.termino).replace("%20", "+")
        if pagina == 1:
            return f"{self.BASE_URL}/?s={termino_encoded}"
        return f"{self.BASE_URL}/page/{pagina}/?s={termino_encoded}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        """
        Extrae el total de páginas desde <a class="last" title="N">.
        Limita según max_articulos (10 artículos por página).
        """
        try:
            a_last = soup.find("a", class_="last")
            if a_last and a_last.get("title", "").isdigit():
                total_paginas_sitio = int(a_last["title"])
            else:
                # Fallback: buscar el número más alto en page-numbers
                nums = []
                for a in soup.find_all("a", class_="page"):
                    t = a.get_text(strip=True)
                    if t.isdigit():
                        nums.append(int(t))
                total_paginas_sitio = max(nums) if nums else 1

            max_paginas_por_limite = math.ceil(self.max_articulos / 10)
            total_paginas = min(total_paginas_sitio, max_paginas_por_limite)

            print(f"  Total páginas en sitio: {total_paginas_sitio} | "
                  f"Límite: {self.max_articulos} arts → recorriendo {total_paginas} páginas")
            return total_paginas
        except Exception:
            return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """
        Extrae links SOLO desde la columna principal (tdi_96),
        ignorando el sidebar (tdi_99).
        Cachea fecha desde el atributo datetime del <time>.
        """
        links = []

        # Buscar la columna principal por clase parcial 'tdi_96'
        columna_principal = soup.find(
            "div",
            class_=lambda c: c and "tdi_96" in c
        )

        if not columna_principal:
            # Fallback: usar todo el soup si no se encuentra la columna
            columna_principal = soup

        modulos = columna_principal.find_all("div", class_="td-module-container")

        for modulo in modulos:
            # Link desde el título
            h3 = modulo.find("h3", class_="td-module-title")
            if not h3:
                continue

            a = h3.find("a")
            if not a:
                continue

            href = a.get("href", "").strip()
            if not href or self.DOMINIO not in href:
                continue

            # Fecha desde datetime="YYYY-MM-DDTHH:MM:SS-05:00"
            fecha = None
            time_tag = modulo.find("time", class_="td-module-date")
            if time_tag and time_tag.get("datetime"):
                fecha = self._parsear_fecha_trochando(time_tag["datetime"])

            self._fechas_cache[href] = fecha
            links.append(href)

        return links

    # ── Override de scrape() con retry + backoff ────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  Límite de artículos: {self.max_articulos}")

        try:
            url_p1 = self._construir_url_busqueda(pagina=1)
            r = self.session.get(url_p1, timeout=20)
            r.raise_for_status()
            soup_p1 = BeautifulSoup(r.text, "html.parser")
        except Exception as e:
            print(f"  Error al cargar primera página: {e}")
            return pd.DataFrame()

        total_paginas = self._obtener_total_paginas(soup_p1)
        links_en_rango = []
        errores_consecutivos = 0

        for pagina in range(1, total_paginas + 1):
            try:
                if pagina == 1:
                    soup = soup_p1
                else:
                    url = self._construir_url_busqueda(pagina=pagina)

                    respuesta = None
                    for intento in range(3):
                        try:
                            r = self.session.get(url, timeout=25)
                            r.raise_for_status()
                            respuesta = r
                            break
                        except Exception:
                            espera = 5 * (intento + 1)
                            print(f"  Intento {intento+1}/3 fallido en página {pagina} — esperando {espera}s...")
                            time.sleep(espera)

                    if respuesta is None:
                        errores_consecutivos += 1
                        print(f"  Página {pagina} descartada tras 3 intentos.")
                        if errores_consecutivos >= 5:
                            print("  5 páginas consecutivas fallidas — deteniendo scraping.")
                            break
                        continue

                    errores_consecutivos = 0
                    soup = BeautifulSoup(respuesta.text, "html.parser")

                links_pagina = self._extraer_links_pagina(soup)
                print(f"  Página {pagina}/{total_paginas}: {len(links_pagina)} artículos")

                for link in links_pagina:
                    fecha = self._fechas_cache.get(link)
                    if fecha is None:
                        links_en_rango.append(link)
                        continue
                    if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                        links_en_rango.append(link)

                time.sleep(1.5)

            except Exception as e:
                print(f"  Error en página {pagina}: {e}")
                errores_consecutivos += 1
                if errores_consecutivos >= 5:
                    print("  5 páginas consecutivas fallidas — deteniendo scraping.")
                    break
                continue

        links_en_rango = list(dict.fromkeys(links_en_rango))
        print(f"  Total links únicos en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Override de _descargar_articulo_individual (newspaper4k) ────────────

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas.get(link, article.publish_date) if hasattr(self, '_fechas') else article.publish_date,
                "texto": article.text
            }
        except Exception as e:
            print(f'Error descargando: {e}')
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50) # 50 concurrent downloads
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

#### **Enlace Television**

```python
class ScraperEnlaceTelevision(ScraperPeriodico):
    """
    Scraper para Enlace Televisión (enlacetelevision.com).

    - Scroll infinito con botón "More Posts" → requiere Playwright
    - Sin paginación por URL ni total de resultados
    - Links en <li class="mvp-blog-story-wrap"> > <a rel="bookmark">
    - Fecha relativa en <span class="mvp-cd-date">: "4 meses ago", "2 días ago"
      → se calcula restando el tiempo a datetime.now()
    - Parada anticipada: cuando todas las fechas de una carga son anteriores
      a fecha_desde, se detiene el scraping
    - max_articulos para limitar resultados (default: 300)
    """

    BASE_URL   = "https://enlacetelevision.com"
    SEARCH_URL = "https://enlacetelevision.com/?s={termino}"
    DOMINIO    = "enlacetelevision.com"
    REGION = "nororiente"

    def __init__(
        self,
        termino: str,
        fecha_desde: str,
        fecha_hasta: str,
        max_articulos: int = 300,
    ):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_articulos = max_articulos
        self._fecha_desde_dt = datetime.strptime(fecha_desde, "%Y-%m-%d")
        self._fecha_hasta_dt = datetime.strptime(fecha_hasta, "%Y-%m-%d")
        self._fechas_cache: Dict[str, Optional[datetime]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Enlace Television"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        termino_encoded = requests.utils.quote(self.termino).replace("%20", "+")
        return f"{self.BASE_URL}/?s={termino_encoded}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        return 1  # No aplica — manejado por Playwright

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """
        Extrae links desde <li class="mvp-blog-story-wrap"> > <a rel="bookmark">.
        Cachea fecha aproximada calculada desde el texto relativo.
        """
        links = []

        items = soup.find_all("li", class_="mvp-blog-story-wrap")

        for item in items:
            a = item.find("a", rel="bookmark")
            if not a:
                continue

            href = a.get("href", "").strip()
            if not href or self.DOMINIO not in href:
                continue

            # Fecha relativa desde <span class="mvp-cd-date">
            fecha = None
            span_fecha = item.find("span", class_="mvp-cd-date")
            if span_fecha:
                texto = span_fecha.get_text(strip=True)
                fecha = self._parsear_fecha_relativa(texto)

            self._fechas_cache[href] = fecha
            links.append(href)

        return links

    # ── Parseo de fecha relativa ────────────────────────────────────────────

    def _parsear_fecha_relativa(self, texto: str) -> Optional[datetime]:
        """
        Parsea fechas relativas del tipo '4 meses ago', '2 días ago',
        '1 año ago', '3 semanas ago', '5 horas ago'.
        Retorna datetime aproximado restando el tiempo a datetime.now().
        """
        if not texto:
            return None

        texto = texto.lower().strip()
        ahora = datetime.now()

        try:
            partes = texto.split()
            if len(partes) < 2:
                return None

            numero = int(partes[0])
            unidad = partes[1]

            if 'año' in unidad or 'ano' in unidad:
                return ahora - timedelta(days=numero * 365)
            elif 'mes' in unidad:
                return ahora - timedelta(days=numero * 30)
            elif 'semana' in unidad:
                return ahora - timedelta(weeks=numero)
            elif 'día' in unidad or 'dia' in unidad:
                return ahora - timedelta(days=numero)
            elif 'hora' in unidad:
                return ahora - timedelta(hours=numero)
            elif 'minuto' in unidad:
                return ahora - timedelta(minutes=numero)
            else:
                return None
        except (ValueError, IndexError):
            return None

    # ── Playwright: scroll infinito con "More Posts" ────────────────────────

    def _get_links_con_playwright(self) -> List[str]:
        import nest_asyncio
        nest_asyncio.apply()

        async def _fetch():
            todos_links   = []
            links_vistos  = set()
            recolectados  = 0
            url_busqueda  = self._construir_url_busqueda()

            async with async_playwright() as p:
                browser = await p.chromium.launch(headless=True)
                page    = await browser.new_page()

                print(f"  Cargando: {url_busqueda}")
                await page.goto(url_busqueda, wait_until="networkidle", timeout=30000)
                await page.wait_for_selector("li.mvp-blog-story-wrap", timeout=15000)

                carga_num = 1
                while recolectados < self.max_articulos:
                    html = await page.content()
                    soup = BeautifulSoup(html, "html.parser")

                    links_pagina = self._extraer_links_pagina(soup)

                    links_nuevos = [l for l in links_pagina if l not in links_vistos]
                    links_vistos.update(links_nuevos)

                    print(f"  Carga {carga_num}: {len(links_nuevos)} artículos nuevos")

                    # Parada anticipada — corregido: comparar datetime con datetime
                    if links_nuevos:
                        fechas_nuevas = [
                            self._fechas_cache.get(l)
                            for l in links_nuevos
                            if self._fechas_cache.get(l) is not None
                        ]
                        if fechas_nuevas and all(f < self._fecha_desde_dt for f in fechas_nuevas):
                            print("  Todos los artículos nuevos son anteriores a fecha_desde — deteniendo.")
                            todos_links.extend(links_nuevos)
                            break

                    todos_links.extend(links_nuevos)
                    recolectados = len(todos_links)

                    if recolectados >= self.max_articulos:
                        print(f"  Límite de {self.max_articulos} artículos alcanzado.")
                        break

                    btn = page.locator("a.mvp-inf-more-but")
                    count = await btn.count()
                    if count == 0:
                        print("  Botón 'More Posts' no encontrado — fin del contenido.")
                        break

                    is_visible = await btn.first.is_visible()
                    if not is_visible:
                        print("  Botón 'More Posts' oculto — fin del contenido.")
                        break

                    await btn.first.scroll_into_view_if_needed()
                    await btn.first.click()
                    await page.wait_for_load_state("networkidle", timeout=15000)
                    await asyncio.sleep(1.5)
                    carga_num += 1

                await browser.close()
            return todos_links

        loop = asyncio.get_event_loop()
        return loop.run_until_complete(_fetch())

    # ── Override de scrape() ────────────────────────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  Límite de artículos: {self.max_articulos}")

        try:
            todos_links = self._get_links_con_playwright()
        except Exception as e:
            print(f"  Error en Playwright: {e}")
            return pd.DataFrame()

        links_unicos = list(dict.fromkeys(todos_links))
        print(f"  Links únicos recolectados: {len(links_unicos)}")

        # Filtrar por rango de fechas — corregido: comparar datetime con datetime
        links_en_rango = []
        for link in links_unicos:
            fecha = self._fechas_cache.get(link)
            if fecha is None:
                links_en_rango.append(link)
                continue
            if self._fecha_desde_dt <= fecha <= self._fecha_hasta_dt:
                links_en_rango.append(link)

        print(f"  Links en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Descargas asíncronas ────────────────────────────────────────────────

    async def _descargar_articulo_async(self, session: aiohttp.ClientSession, info: tuple) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text
            }
        except Exception as e:
            print(f'Error descargando: {e}')
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [self._descargar_articulo_async(session, info) for info in links_enumerados]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(f"{len(data)} artículos descargados exitosamente ({len(data)/len(links)*100:.1f}%)")
        return pd.DataFrame(data)
```

#### **Corrillos**

```python
class ScraperCorrillos(ScraperPeriodico):
    """
    Scraper para Corrillos (corrillos.com.co).

    - HTML estático: requests + BeautifulSoup (sin Playwright)
    - Paginación por URL: /page/N/?s=termino
    - Total de páginas desde el último <a class="page-numbers"> con número
    - 10 artículos por página
    - Links en <h2 class="post-title theme_blog_post__Title"> > <a>
    - Solo dentro de <div class="col-lg-8 col-md-12"> (columna principal)
    - Fecha en <span class="post-meta-date post_post_item_Date">: "agosto 13, 2024"
    - Formato de fecha de salida: "YYYY-MM-DD"
    - max_articulos para limitar resultados (default: 300)
    """

    BASE_URL = "https://www.corrillos.com.co"
    DOMINIO  = "corrillos.com.co"

    HEADERS = {
        "User-Agent": (
            "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
            "AppleWebKit/537.36 (KHTML, like Gecko) "
            "Chrome/120.0.0.0 Safari/537.36"
        ),
        "Referer": "https://www.corrillos.com.co/",
    }

    MESES_ES = {
        "enero": 1, "febrero": 2, "marzo": 3, "abril": 4,
        "mayo": 5, "junio": 6, "julio": 7, "agosto": 8,
        "septiembre": 9, "octubre": 10, "noviembre": 11, "diciembre": 12,
    }

    REGION = "nororiente"

    def __init__(
        self,
        termino: str,
        fecha_desde: str,
        fecha_hasta: str,
        max_articulos: int = 300,
    ):
        super().__init__(termino, fecha_desde, fecha_hasta)
        self.max_articulos = max_articulos
        self.session.headers.update(self.HEADERS)
        self._fechas_cache: Dict[str, Optional[str]] = {}

    # ── Contrato abstracto ──────────────────────────────────────────────────

    @property
    def nombre_periodico(self) -> str:
        return "Corrillos"

    def _construir_url_busqueda(self, pagina: int = 1) -> str:
        termino_encoded = requests.utils.quote(self.termino).replace("%20", "+")
        if pagina == 1:
            return f"{self.BASE_URL}/?s={termino_encoded}"
        return f"{self.BASE_URL}/page/{pagina}/?s={termino_encoded}"

    def _obtener_total_paginas(self, soup: BeautifulSoup) -> int:
        """
        Extrae el total de páginas desde el mayor número en <a class="page-numbers">.
        Limita según max_articulos (10 artículos por página).
        """
        try:
            numeros = []
            for a in soup.find_all("a", class_="page-numbers"):
                texto = a.get_text(strip=True)
                if texto.isdigit():
                    numeros.append(int(texto))

            total_paginas_sitio = max(numeros) if numeros else 1
            max_paginas_por_limite = math.ceil(self.max_articulos / 10)
            total_paginas = min(total_paginas_sitio, max_paginas_por_limite)

            print(f"  Total páginas en sitio: {total_paginas_sitio} | "
                  f"Límite: {self.max_articulos} arts → recorriendo {total_paginas} páginas")
            return total_paginas
        except Exception:
            return 1

    def _extraer_links_pagina(self, soup: BeautifulSoup) -> List[str]:
        """
        Extrae links SOLO desde <div class="col-lg-8 col-md-12"> (columna principal).
        Cachea fecha desde <span class="post-meta-date post_post_item_Date">.
        """
        links = []

        columna = soup.find("div", class_=lambda c: c and "col-lg-8" in c and "col-md-12" in c)
        if not columna:
            columna = soup

        articulos = columna.find_all("article")

        for articulo in articulos:
            h2 = articulo.find("h2", class_="post-title")
            if not h2:
                continue

            a = h2.find("a")
            if not a:
                continue

            href = a.get("href", "").strip()
            if not href or self.DOMINIO not in href:
                continue

            fecha = None
            span_fecha = articulo.find("span", class_="post-meta-date")
            if span_fecha:
                texto = span_fecha.get_text(strip=True)
                fecha = self._parsear_fecha_corrillos(texto)

            self._fechas_cache[href] = fecha
            links.append(href)

        return links

    # ── Parseo de fecha ────────────────────────────────────────────────────

    def _parsear_fecha_corrillos(self, texto: str) -> Optional[str]:
        """
        Parsea fechas en formato español: "agosto 13, 2024" o "agosto 13 de 2024".
        Retorna string "YYYY-MM-DD" o None si no puede parsear.
        """
        if not texto:
            return None

        texto = texto.lower().strip()

        # Limpiar " de " entre día y año
        texto = texto.replace(" de ", " ")

        # Quitar comas
        texto = texto.replace(",", "")

        partes = texto.split()
        if len(partes) < 3:
            return None

        try:
            mes_str = partes[0]
            dia     = int(partes[1])
            anio    = int(partes[2])
            mes     = self.MESES_ES.get(mes_str)

            if mes is None:
                return None

            return f"{anio:04d}-{mes:02d}-{dia:02d}"
        except (ValueError, IndexError):
            return None

    # ── Override de scrape() con retry + backoff ────────────────────────────

    def scrape(self) -> pd.DataFrame:
        print(f"WEBSCRAPING {self.nombre_periodico.upper()}")
        print(f"  Término: '{self.termino}' | Rango: {self.fecha_desde} → {self.fecha_hasta}")
        print(f"  Límite de artículos: {self.max_articulos}")

        try:
            url_p1 = self._construir_url_busqueda(pagina=1)
            r = self.session.get(url_p1, timeout=20)
            r.raise_for_status()
            soup_p1 = BeautifulSoup(r.text, "html.parser")
        except Exception as e:
            print(f"  Error al cargar primera página: {e}")
            return pd.DataFrame()

        total_paginas = self._obtener_total_paginas(soup_p1)
        links_en_rango = []
        errores_consecutivos = 0

        for pagina in range(1, total_paginas + 1):
            try:
                if pagina == 1:
                    soup = soup_p1
                else:
                    url = self._construir_url_busqueda(pagina=pagina)

                    respuesta = None
                    for intento in range(3):
                        try:
                            r = self.session.get(url, timeout=25)
                            r.raise_for_status()
                            respuesta = r
                            break
                        except Exception:
                            espera = 5 * (intento + 1)
                            print(f"  Intento {intento+1}/3 fallido en página {pagina} "
                                  f"— esperando {espera}s...")
                            time.sleep(espera)

                    if respuesta is None:
                        errores_consecutivos += 1
                        print(f"  Página {pagina} descartada tras 3 intentos.")
                        if errores_consecutivos >= 5:
                            print("  5 páginas consecutivas fallidas — deteniendo.")
                            break
                        continue

                    errores_consecutivos = 0
                    soup = BeautifulSoup(respuesta.text, "html.parser")

                links_pagina = self._extraer_links_pagina(soup)
                print(f"  Página {pagina}/{total_paginas}: {len(links_pagina)} artículos")

                for link in links_pagina:
                    fecha = self._fechas_cache.get(link)
                    if fecha is None:
                        links_en_rango.append(link)
                        continue
                    if self.fecha_desde <= fecha <= self.fecha_hasta:
                        links_en_rango.append(link)

                time.sleep(1.0)

            except Exception as e:
                print(f"  Error en página {pagina}: {e}")
                errores_consecutivos += 1
                if errores_consecutivos >= 5:
                    print("  5 páginas consecutivas fallidas — deteniendo.")
                    break
                continue

        links_en_rango = list(dict.fromkeys(links_en_rango))
        print(f"  Total links únicos en rango de fechas: {len(links_en_rango)}")

        if not links_en_rango:
            print("  No se encontraron artículos en el rango de fechas indicado.")
            return pd.DataFrame()

        return self._descargar_articulos(links_en_rango)

    # ── Descargas asíncronas ────────────────────────────────────────────────

    async def _descargar_articulo_async(
        self, session: aiohttp.ClientSession, info: tuple
    ) -> Optional[Dict]:
        i, link = info
        try:
            async with session.get(link, timeout=15) as response:
                html = await response.text()

            from newspaper import Article
            article = Article(link)
            article.html = html
            article.download_state = 2
            await asyncio.to_thread(article.parse)

            if i % 50 == 0:
                print(f"  Descargados: {i} artículos")

            return {
                "periodico": self.nombre_periodico,
                "url": link,
                "titulo": article.title,
                # corregido: usar _fechas_cache en vez de _fechas
                "fecha": self._fechas_cache.get(link, article.publish_date),
                "texto": article.text,
            }
        except Exception as e:
            print(f"Error descargando: {e}")
            return None

    async def _ejecutar_descargas_async(self, links: List[str]):
        links_enumerados = [(i, link) for i, link in enumerate(links, 1)]
        connector = aiohttp.TCPConnector(limit=50)
        async with aiohttp.ClientSession(connector=connector) as session:
            tareas = [
                self._descargar_articulo_async(session, info)
                for info in links_enumerados
            ]
            resultados = await asyncio.gather(*tareas)
        return resultados

    def _descargar_articulos(self, links: List[str]) -> pd.DataFrame:
        print(f"\nDescargando {len(links)} artículos (Optimizados con Asincronía)...")
        if not links:
            return pd.DataFrame()

        try:
            loop = asyncio.get_event_loop()
        except RuntimeError:
            loop = asyncio.new_event_loop()
            asyncio.set_event_loop(loop)

        resultados = loop.run_until_complete(self._ejecutar_descargas_async(links))

        data = [r for r in resultados if r is not None]
        print(
            f"{len(data)} artículos descargados exitosamente "
            f"({len(data)/len(links)*100:.1f}%)"
        )
        return pd.DataFrame(data)
```

```python
class GestorScraping:
    """Gestiona el webscraping de múltiples periódicos"""

    SCRAPERS = {
        # Eje Cafetero + Antioquia
        'eldiario': ScraperElDiario,
        'elcolombiano': ScraperElColombiano,
        'bcnoticias': ScraperBCNoticias,
        'elquindiano': ScraperElQuindiano,
        # Suroccidente (Pacífico)
        'elpais': ScraperElPaisCali,
        'diariooccidente': ScraperDiarioOccidente,
        'diariodelsur': ScraperDiarioDelSur,
        'diariodelcauca': ScraperDiarioDelCauca,
        'choco7dias': ScraperChoco7Dias,
        # Orinoquía / Amazonía
        'llanoalmundo': ScraperLlanoAlMundo,
        'diariodecasanare': ScraperDiarioDeCasanare,
        'lavozdelcinaruco': ScraperLaVozDelCinaruco,
        'elmorichal': ScraperElMorichal,
        'miputumayo': ScraperMiPutumayo,
        # Central (Bogotá)
        'eltiempo': ScraperElTiempo,
        'larepublica': ScraperLaRepublica,
        'portafolio': ScraperPortafolio,
        'publimetro': ScraperPublimetro,
        'las2orillas': ScraperLas2Orillas,
        # Caribe
        'elheraldo': ScraperElHeraldo,
        'eluniversal': ScraperElUniversal,
        'elpilon': ScraperElPilon,
        'elmeridiano': ScraperElMeridiano,
        # Nororiente
        'vanguardia': ScraperVanguardia,
        'tronchandosinfronteras': ScraperTrochandoSinFronteras,
        'enlacetelevision': ScraperEnlaceTelevision,
        'corrillos': ScraperCorrillos,
    }

    def __init__(self, termino: str, fecha_desde: str, fecha_hasta: str):
        self.termino = termino
        self.fecha_desde = fecha_desde
        self.fecha_hasta = fecha_hasta

    @classmethod
    def periodicos_disponibles(cls) -> List[str]:
        return list(cls.SCRAPERS.keys())

    # ── FILTRO DE RELEVANCIA ──────────────────────────────────────────────────

    def _es_relevante(self, texto: str, titulo: str,
                      min_menciones: int = 3) -> bool:
        territorio = self.termino.split()[0].lower()
        texto_lower  = (texto  or "").lower()
        titulo_lower = (titulo or "").lower()
        menciones_texto = texto_lower.count(territorio)
        bonus_titulo    = 2 if territorio in titulo_lower else 0
        return (menciones_texto + bonus_titulo) >= min_menciones

    def _filtrar_relevancia(self, df: pd.DataFrame,
                             min_menciones: int = 3) -> pd.DataFrame:
        if df.empty:
            return df
        mascara = df.apply(
            lambda row: self._es_relevante(
                row.get('texto', ''),
                row.get('titulo', ''),
                min_menciones,
            ),
            axis=1
        )
        df_filtrado = df[mascara].reset_index(drop=True)
        descartados = len(df) - len(df_filtrado)
        print(f"   Filtro relevancia ({self.termino.split()[0]}, "
              f"umbral={min_menciones}): "
              f"{len(df)} → {len(df_filtrado)} artículos "
              f"({descartados} descartados)")
        return df_filtrado

    # ─────────────────────────────────────────────────────────────────────────

    def scrape_multiples(self, periodicos: List[str],
                         callback=None,
                         min_menciones: int = 3) -> pd.DataFrame:
        dataframes = []
        for periodico_id in periodicos:
            if periodico_id not in self.SCRAPERS:
                print(f"Periódico '{periodico_id}' no disponible.")
                continue
            scraper_class = self.SCRAPERS[periodico_id]
            try:
                scraper = scraper_class(self.termino, self.fecha_desde, self.fecha_hasta)
                df = scraper.scrape()
                if not df.empty and min_menciones > 0:
                    df = self._filtrar_relevancia(df, min_menciones)
                if not df.empty:
                    dataframes.append(df)
                if callback:
                    callback(periodico_id, df)
            except Exception as e:
                print(f"Error scraping {periodico_id}: {e}")
                if callback:
                    callback(periodico_id, pd.DataFrame())
        if dataframes:
            df_final = self._unificar_dataframes(dataframes)
            self._mostrar_resumen(df_final)
            return df_final
        else:
            print("No se pudo obtener ningún artículo")
            return pd.DataFrame(columns=['periodico', 'titulo', 'fecha', 'texto', 'url'])

    def _unificar_dataframes(self, dataframes: List[pd.DataFrame]) -> pd.DataFrame:
        df_final = pd.concat(dataframes, ignore_index=True)
        columnas_requeridas = ['periodico', 'titulo', 'fecha', 'texto', 'url']
        for col in columnas_requeridas:
            if col not in df_final.columns:
                df_final[col] = None
        df_final = df_final[columnas_requeridas]
        df_final['fecha'] = pd.to_datetime(df_final['fecha'], errors='coerce', utc=True)
        df_final['fecha'] = df_final['fecha'].dt.tz_localize(None)
        df_final = df_final.sort_values('fecha', ascending=False).reset_index(drop=True)
        return df_final

    def _mostrar_resumen(self, df: pd.DataFrame):
        print(f"\n{'='*50}")
        print(f"RESUMEN FINAL")
        print(f"{'='*50}")
        print(f"Total de artículos: {len(df)}")
        print(f"\nPor periódico:")
        print(df['periodico'].value_counts().to_string())
        if df['fecha'].notna().any():
            print(f"\nRango de fechas:")
            print(f"  Desde: {df['fecha'].min():%Y-%m-%d}")
            print(f"  Hasta: {df['fecha'].max():%Y-%m-%d}")
        print(f"{'='*50}\n")

    @classmethod
    def regiones_disponibles(cls) -> List[str]:
        regiones = set()
        for scraper_class in cls.SCRAPERS.values():
            region = getattr(scraper_class, "REGION", None)
            if region:
                regiones.add(region)
        return sorted(regiones)

    @classmethod
    def periodicos_por_region(cls, region: str) -> List[str]:
        return [
            clave
            for clave, scraper_class in cls.SCRAPERS.items()
            if getattr(scraper_class, "REGION", None) == region
        ]


# ══════════════════════════════════════════════════════════════════════════════
# FUNCIONES DE BÚSQUEDA
# ══════════════════════════════════════════════════════════════════════════════

def scrape_periodicos(termino: str, fecha_desde: str, fecha_hasta: str,
                      periodicos: List[str] = None,
                      region: str = None,
                      min_menciones: int = 3) -> pd.DataFrame:
    """Función base para webscraping con filtro de relevancia integrado."""
    gestor = GestorScraping(termino, fecha_desde, fecha_hasta)
    if periodicos is not None:
        lista_final = periodicos
    elif region is not None:
        lista_final = gestor.periodicos_por_region(region)
        if not lista_final:
            print(f"No se encontraron periódicos para la región '{region}'.")
            return pd.DataFrame(columns=['periodico', 'titulo', 'fecha', 'texto', 'url'])
        print(f"Región '{region}': {len(lista_final)} periódicos → {lista_final}")
    else:
        lista_final = gestor.periodicos_disponibles()
    return gestor.scrape_multiples(lista_final, min_menciones=min_menciones)


def _buscar_multi_termino(territorio: str, fecha_desde: str, fecha_hasta: str,
                          periodicos: List[str], min_menciones: int,
                          temas: List[str]) -> pd.DataFrame:
    """
    Función interna compartida por scrape_municipio y scrape_departamento.
    Realiza búsquedas por múltiples temas, deduplica por URL y agrega
    la columna 'terminos_encontrado'.
    """
    print(f"\n{'='*60}")
    print(f"BUSQUEDA MULTI-TERMINO: {territorio} | {fecha_desde} -> {fecha_hasta}")
    print(f"Terminos: {[f'{territorio} {t}' for t in temas]}")
    print(f"{'='*60}")

    resultados_por_termino = {}

    for tema in temas:
        termino = f"{territorio} {tema}"
        print(f"\n-- Buscando: '{termino}' --")
        df_tema = scrape_periodicos(
            termino=termino,
            fecha_desde=fecha_desde,
            fecha_hasta=fecha_hasta,
            periodicos=periodicos,
            min_menciones=min_menciones,
        )
        if not df_tema.empty:
            df_tema['terminos_encontrado'] = tema
            resultados_por_termino[tema] = df_tema
            print(f"   -> {len(df_tema)} articulos relevantes")
        else:
            print(f"   -> Sin resultados")

    if not resultados_por_termino:
        return pd.DataFrame()

    df_todos = pd.concat(resultados_por_termino.values(), ignore_index=True)

    terminos_por_url = (
        df_todos.groupby('url')['terminos_encontrado']
        .apply(lambda x: ', '.join(sorted(set(x))))
        .reset_index()
    )

    df_dedup = (
        df_todos
        .drop(columns=['terminos_encontrado'])
        .drop_duplicates(subset=['url'], keep='first')
        .merge(terminos_por_url, on='url', how='left')
        .sort_values('fecha', ascending=False)
        .reset_index(drop=True)
    )

    total_antes = len(df_todos)
    total_dedup = len(df_dedup)
    print(f"\n{'='*60}")
    print(f"RESUMEN -- {territorio}")
    print(f"{'='*60}")
    for tema, df_t in resultados_por_termino.items():
        print(f"  {territorio} {tema:<15}: {len(df_t):>4}")
    print(f"{'─'*40}")
    print(f"Total antes de deduplicar : {total_antes:>4}")
    print(f"Duplicados eliminados     : {total_antes - total_dedup:>4}")
    print(f"Articulos unicos finales  : {total_dedup:>4}")
    print(f"\nCobertura por termino:")
    cobertura = df_dedup['terminos_encontrado'].str.split(', ').explode().value_counts()
    for tema, n in cobertura.items():
        print(f"  {tema:<15}: {n}")
    if df_dedup['fecha'].notna().any():
        print(f"\nRango de fechas: {df_dedup['fecha'].min():%Y-%m-%d} -> "
              f"{df_dedup['fecha'].max():%Y-%m-%d}")
    print(f"{'='*60}\n")

    return df_dedup


# Temas de búsqueda: cubren las 5 dimensiones del pipeline
_TEMAS = ["conflicto", "comunidades", "institucional", "derechos", "social"]

# Periódicos centrales de respaldo cuando el corpus local es insuficiente
_PERIODICOS_RESPALDO = ['eltiempo', 'las2orillas']

# Umbral mínimo de artículos antes de activar el respaldo
_MIN_ARTICULOS_RESPALDO = 50


def scrape_municipio(municipio: str, fecha_desde: str, fecha_hasta: str,
                     periodicos: List[str] = None,
                     region: str = None,
                     min_menciones: int = 3) -> pd.DataFrame:
    """
    Busca artículos de un municipio usando 5 términos temáticos y
    deduplica por URL. Agrega columna 'terminos_encontrado'.
    """
    if periodicos is None and region is not None:
        gestor = GestorScraping.__new__(GestorScraping)
        periodicos = gestor.periodicos_por_region(region)

    df = _buscar_multi_termino(municipio, fecha_desde, fecha_hasta,
                                periodicos, min_menciones, _TEMAS)
    if not df.empty:
        df['municipio'] = municipio
    return df


# ── MAPEO DEPARTAMENTO → PERIÓDICOS ──────────────────────────────────────────

DEPARTAMENTO_PERIODICOS = {
    'Antioquia':       ['elcolombiano'],
    'Risaralda':       ['eldiario'],
    'Caldas':          ['bcnoticias'],
    'Quindío':         ['elquindiano'],
    'Valle del Cauca': ['elpais', 'diariooccidente'],
    'Nariño':          ['diariodelsur'],
    'Cauca':           ['diariodelcauca'],
    'Chocó':           ['choco7dias'],
    'Meta':            ['llanoalmundo'],
    'Casanare':        ['diariodecasanare'],
    'Arauca':          ['lavozdelcinaruco', 'tronchandosinfronteras'],
    'Vichada':         ['elmorichal'],
    'Putumayo':        ['miputumayo'],
    'Cundinamarca':    ['eltiempo', 'larepublica', 'portafolio', 'publimetro', 'las2orillas'],
    'Atlántico':       ['elheraldo'],
    'Bolívar':         ['eluniversal'],
    'Cesar':           ['elpilon'],
    'Córdoba':         ['elmeridiano'],
    'Santander':       ['vanguardia', 'enlacetelevision', 'corrillos'],
}


def scrape_departamento(departamento: str,
                        fecha_desde: str,
                        fecha_hasta: str,
                        periodicos: List[str] = None,
                        min_menciones: int = None,
                        min_articulos: int = _MIN_ARTICULOS_RESPALDO,
                        usar_respaldo: bool = True) -> pd.DataFrame:
    """
    Busca artículos de un departamento usando 5 términos temáticos y
    deduplica por URL. Usa automáticamente los scrapers del departamento
    según DEPARTAMENTO_PERIODICOS.

    El umbral de menciones se determina automáticamente por departamento
    usando DEPARTAMENTO_MIN_MENCIONES (3 para departamentos grandes,
    2 para pequeños), pero puede sobreescribirse con min_menciones.

    Si el corpus resultante tiene menos de `min_articulos`, activa búsqueda
    de respaldo en El Tiempo y Las2Orillas (cobertura nacional).

    Args:
        departamento:  Nombre del departamento (ej: 'Antioquia')
        fecha_desde:   Fecha inicio (YYYY-MM-DD)
        fecha_hasta:   Fecha fin (YYYY-MM-DD)
        periodicos:    Lista de claves de scrapers. Si es None, usa los del
                       mapeo DEPARTAMENTO_PERIODICOS.
        min_menciones: Mínimo de menciones del departamento en el texto.
                       Si es None, usa el valor de DEPARTAMENTO_MIN_MENCIONES.
                       Default: None (automático).
        min_articulos: Umbral mínimo. Si el corpus queda por debajo,
                       se activa el respaldo. Default: 50.
        usar_respaldo: Activar búsqueda en periódicos centrales si el
                       corpus local es insuficiente. Default: True.

    Returns:
        DataFrame deduplicado con columnas estándar + 'departamento' +
        'terminos_encontrado'.
    """
    # Determinar umbral automáticamente si no se pasa explícitamente
    if min_menciones is None:
        min_menciones = DEPARTAMENTO_MIN_MENCIONES.get(departamento, 3)
        print(f"  Umbral de menciones para {departamento}: {min_menciones} (automático)")

    if periodicos is None:
        periodicos = DEPARTAMENTO_PERIODICOS.get(departamento)
        if periodicos is None:
            print(f"  Advertencia: '{departamento}' no está en el mapeo. "
                  f"Usando todos los scrapers disponibles.")
            periodicos = GestorScraping.periodicos_disponibles()
        else:
            print(f"  Periódicos para {departamento}: {periodicos}")

    # Búsqueda principal en periódicos locales
    df = _buscar_multi_termino(departamento, fecha_desde, fecha_hasta,
                                periodicos, min_menciones, _TEMAS)

    # Respaldo automático si el corpus es insuficiente
    if usar_respaldo and len(df) < min_articulos:
        # No repetir scrapers que ya se usaron
        respaldo_disponible = [
            p for p in _PERIODICOS_RESPALDO if p not in periodicos
        ]

        if respaldo_disponible:
            print(f"\n  Solo {len(df)} artículos para {departamento} "
                  f"(mínimo recomendado: {min_articulos})")
            print(f"  Activando búsqueda de respaldo en: {respaldo_disponible}")

            df_respaldo = _buscar_multi_termino(
                departamento, fecha_desde, fecha_hasta,
                respaldo_disponible, min_menciones, _TEMAS
            )

            if not df_respaldo.empty:
                df = (
                    pd.concat([df, df_respaldo], ignore_index=True)
                    .drop_duplicates(subset=['url'])
                    .sort_values('fecha', ascending=False)
                    .reset_index(drop=True)
                )
                print(f"  ✓ Total con respaldo: {len(df)} artículos")
            else:
                print(f"  Sin resultados adicionales en periódicos de respaldo.")

    if not df.empty:
        df['departamento'] = departamento
    else:
        return pd.DataFrame(columns=[
            'periodico', 'titulo', 'fecha', 'texto', 'url',
            'departamento', 'terminos_encontrado'
        ])

    return df


def scrape_multiples_departamentos(departamentos: List[str],
                                   fecha_desde: str,
                                   fecha_hasta: str,
                                   min_menciones: int = 3,
                                   min_articulos: int = _MIN_ARTICULOS_RESPALDO,
                                   usar_respaldo: bool = True) -> pd.DataFrame:
    """
    Corre scrape_departamento() para una lista de departamentos y
    concatena los resultados en un solo DataFrame.

    Args:
        departamentos: Lista de nombres de departamentos
        fecha_desde:   Fecha inicio (YYYY-MM-DD)
        fecha_hasta:   Fecha fin (YYYY-MM-DD)
        min_menciones: Umbral de relevancia. Default: 3.
        min_articulos: Umbral mínimo para activar respaldo. Default: 50.
        usar_respaldo: Activar búsqueda en periódicos centrales si hay
                       pocos artículos locales. Default: True.

    Returns:
        DataFrame unificado con todos los artículos, columna 'departamento'
        incluida.
    """
    resultados = []

    for dep in departamentos:
        print(f"\n{'#'*60}")
        print(f"# PROCESANDO: {dep}")
        print(f"{'#'*60}")
        df_dep = scrape_departamento(
            departamento=dep,
            fecha_desde=fecha_desde,
            fecha_hasta=fecha_hasta,
            min_menciones=min_menciones,
            min_articulos=min_articulos,
            usar_respaldo=usar_respaldo,
        )
        if not df_dep.empty:
            resultados.append(df_dep)

    if not resultados:
        print("Sin artículos en ningún departamento.")
        return pd.DataFrame()

    df_final = pd.concat(resultados, ignore_index=True)

    print(f"\n{'='*60}")
    print(f"RESUMEN TOTAL — {len(departamentos)} departamentos")
    print(f"{'='*60}")
    print(f"Total artículos únicos: {len(df_final)}")
    print(f"\nPor departamento:")
    print(df_final['departamento'].value_counts().to_string())
    print(f"{'='*60}\n")

    return df_final


# ── UMBRAL DE MENCIONES RECOMENDADO POR DEPARTAMENTO ─────────────────────────
# Departamentos grandes con alto volumen periodístico: umbral 3
# Departamentos pequeños o con baja cobertura mediática: umbral 2

DEPARTAMENTO_MIN_MENCIONES = {
    # Umbral 3 — departamentos grandes / buena cobertura
    'Antioquia':       3,
    'Cundinamarca':    3,
    'Valle del Cauca': 3,
    'Atlántico':       3,
    'Bolívar':         3,
    'Santander':       3,
    # Umbral 2 — departamentos pequeños / baja cobertura mediática
    'Risaralda':       2,
    'Caldas':          2,
    'Quindío':         2,
    'Nariño':          2,
    'Cauca':           2,
    'Chocó':           2,
    'Meta':            2,
    'Casanare':        2,
    'Arauca':          2,
    'Vichada':         2,
    'Putumayo':        2,
    'Cesar':           2,
    'Córdoba':         2,
}
```

## **ClasificadorTematico**

```python
class ClasificadorTemas:
    """
    Clasifica TEMAS ESTRUCTURALES (de qué habla el texto) usando zero-shot.

    Dimensiones asignadas:
    - participacion_comunitaria   → Dim 1: Gobernanza y riesgo institucional (FUSIONAR)
    - incentivos_economicos       → Dim 3: Vulneración socioeconómica (REFORMULAR)
    - fortalecimiento_institucional → Dim 2: Capacidad institucional (MANTENER)
    - impactos_ambientales        → Dim 4: Vulnerabilidad territorial (REFORMULAR)
    - conflictos_socioambientales → Dim 4: Vulnerabilidad territorial (MANTENER)
    """

    def __init__(self,
                 modelo_nli_nombre="MoritzLaurer/mDeBERTa-v3-base-xnli-multilingual-nli-2mil7"):
        print("\nINICIALIZANDO CLASIFICADOR DE TEMAS")
        print(f"Cargando clasificador zero-shot: {modelo_nli_nombre}...")

        self.clasificador = transformers_pipeline(
            "zero-shot-classification",
            model=modelo_nli_nombre,
            device=0 if torch.cuda.is_available() else -1
        )

        print(f"Clasificador cargado")

        # HIPÓTESIS REFORMULADAS en lógica de riesgo/déficit
        # participacion_comunitaria: se fusionará en pipeline con taller, dialogo, audiencia
        # incentivos_economicos: reformulado → concentración desigual de beneficios
        # impactos_ambientales: reformulado → severidad de impactos socioambientales
        self.temas_dict = {
            "participacion_comunitaria": (
                "déficit de participación comunitaria, ausencia de participación ciudadana, "
                "exclusión de veedurías ciudadanas, debilidad del control social"
            ),
            "incentivos_economicos": (
                "concentración desigual de beneficios, distribución inequitativa de incentivos "
                "económicos, captura de rentas, exclusión económica de comunidades locales"
            ),
            # TIGHTENED: requiere debilidad documentada y verificada por ente formal
            "fortalecimiento_institucional": (
                "debilidad institucional grave y documentada: entidades con déficit de recursos "
                "verificado, incapacidad demostrada para cumplir su mandato, carencia de personal "
                "técnico o infraestructura que impide prestar servicios esenciales, reportada "
                "por ente de control o investigación formal"
            ),
            # TIGHTENED: requiere daño medible concreto con consecuencias comunitarias documentadas
            "impactos_ambientales": (
                "daño ambiental concreto y medible con consecuencias directas para comunidades: "
                "contaminación cuantificada de fuentes de agua, deforestación con pérdida de "
                "hectáreas reportada, destrucción de ecosistemas que afecta medios de vida "
                "específicos de poblaciones identificadas"
            ),
            # TIGHTENED: requiere conflicto activo con partes identificadas y acciones concretas
            "conflictos_socioambientales": (
                "conflicto socioambiental activo y en curso entre comunidades y actores "
                "extractivos o estatales por control, uso o daño de recursos naturales o "
                "territorios, con partes identificadas, demandas concretas y acciones "
                "de disputa documentadas"
            ),
        }

        self.temas_labels = list(self.temas_dict.values())
        self.temas_cortos = list(self.temas_dict.keys())

        print(f"Configurado con {len(self.temas_dict)} temas estructurales")
        print()

    def clasificar_texto(self, texto: str, umbral: float = 0.7):
        if pd.isna(texto) or not texto.strip():
            return []

        # multi_label=False: solo el TEMA PRINCIPAL del artículo.
        resultado = self.clasificador(
            texto,
            candidate_labels=self.temas_labels,
            multi_label=False
        )

        # Devuelve TODOS los scores — el umbral lo aplica el pipeline
        # en _consolidar_resultados para categorias_detectadas, no aquí
        return [
            {'tema': label, 'probabilidad': score}
            for label, score in zip(resultado['labels'], resultado['scores'])
        ]

    def clasificar_dataframe(self, df: pd.DataFrame,
                             columna_texto: str = 'texto',
                             umbral: float = 0.7):
        print("CLASIFICACIÓN DE TEMAS ESTRUCTURALES")
        print(f"Total de textos: {len(df)}")
        print(f"Umbral: {umbral}")
        print()

        resultados = []

        for idx, texto in enumerate(df[columna_texto]):
            if pd.isna(texto):
                resultados.append([])
            else:
                temas = self.clasificar_texto(texto, umbral=umbral)
                resultados.append(temas)

            if (idx + 1) % 10 == 0 or (idx + 1) == len(df):
                print(f"   Procesados: {idx + 1}/{len(df)}")

        print()
        # Columnas numéricas (score 0.0–1.0) — no binario
        # 0.0 = no alcanzó el umbral, >0 = probabilidad del tema
        print("   Creando columnas de score numérico...")
        for nombre_corto, label_larga in self.temas_dict.items():
            df[nombre_corto] = [
                next(
                    (t['probabilidad'] for t in resultado if t['tema'] == label_larga),
                    0.0
                )
                for resultado in resultados
            ]

        print("Columnas creadas")
        print()

        return df
```

## **DetectorEventos**

```python
class DetectorEventos:
    """
    Detecta eventos PUNTUALES que OCURRIERON usando NLI + extracción de nuggets.

    Dimensiones asignadas:
    - desplazamiento_forzado  → Dim 5: Derechos humanos y conflicto (MANTENER)
    - reasentamiento          → Dim 4: Vulnerabilidad territorial (MANTENER)
    - protesta_social         → Dim 3: Vulneración socioeconómica (MANTENER)
    - amenaza_intimidacion    → Dim 5: Derechos humanos y conflicto (MANTENER)
    - consulta_previa         → Dim 1: Gobernanza y riesgo institucional (MANTENER)
    - audiencia_publica       → Dim 1: Gobernanza — FUSIONAR en pipeline con
                                participacion_comunitaria, taller_participativo, llamado_dialogo
    - taller_participativo    → Dim 1: Gobernanza — FUSIONAR en pipeline (igual que audiencia)
    - conflicto_territorial   → Dim 4: Vulnerabilidad territorial (MANTENER)
    """

    def __init__(self,
                 modelo_nli_preloaded=None,
                 tokenizer_nli_preloaded=None,
                 modelo_spacy="es_core_news_lg"):
        print("\n INICIALIZANDO DETECTOR DE EVENTOS")

        if modelo_nli_preloaded is None or tokenizer_nli_preloaded is None:
            raise ValueError(
                "DetectorEventos requiere modelo_nli_preloaded y tokenizer_nli_preloaded."
            )

        self.modelo_nli = modelo_nli_preloaded
        self.tokenizer_nli = tokenizer_nli_preloaded
        self.device = next(self.modelo_nli.parameters()).device

        print(f" Modelo NLI compartido conectado (device: {self.device})")

        self.label_entailment = 0
        self.label_neutral = 1
        self.label_contradiction = 2

        print(f"Cargando SpaCy: {modelo_spacy}...")
        try:
            self.nlp = spacy.load(modelo_spacy)
            print(f" SpaCy cargado")
        except OSError:
            print(f"Modelo {modelo_spacy} no encontrado. Usando es_core_news_sm...")
            self.nlp = spacy.load("es_core_news_sm")

        # HIPÓTESIS REFORMULADAS en lógica de riesgo/intensidad
        # consulta_previa: reformulada para capturar déficit/ausencia
        # audiencia_publica y taller_participativo: hipótesis orientadas a
        #   detectar déficit o calidad, se fusionan en el pipeline
        self.eventos_dict = {
            "desplazamiento_forzado": (
                "Este texto reporta que OCURRIÓ un desplazamiento forzado, "
                "expulsión o éxodo de comunidades"
            ),
            "reasentamiento": (
                "Este texto reporta que SE REALIZÓ un reasentamiento, reubicación "
                "o traslado planificado de población"
            ),
            "protesta_social": (
                "Este texto reporta que OCURRIÓ una protesta social, manifestación, "
                "bloqueo, movilización o paro"
            ),
            "amenaza_intimidacion": (
                "Este texto reporta que OCURRIERON amenazas, intimidación, "
                "hostigamiento o violencia contra personas"
            ),
            # TIGHTENED: requiere pueblo étnico identificado + proyecto concreto
            "consulta_previa": (
                "Este texto reporta ESPECÍFICAMENTE la realización o la omisión deliberada "
                "de una consulta previa libre e informada a un pueblo indígena o comunidad "
                "afrodescendiente identificada, en relación con un proyecto o decisión concreta "
                "que afecta su territorio o derechos colectivos"
            ),
            # Se fusionará en pipeline con participacion_comunitaria, taller, llamado_dialogo
            "audiencia_publica": (
                "Este texto reporta que SE REALIZÓ o SE OMITIÓ una audiencia pública, "
                "socialización o reunión informativa oficial"
            ),
            # Se fusionará en pipeline con participacion_comunitaria, audiencia, llamado_dialogo
            "taller_participativo": (
                "Este texto reporta que SE REALIZÓ o SE OMITIÓ un taller participativo, "
                "taller comunitario o sesión de trabajo con comunidades"
            ),
            # TIGHTENED: requiere conflicto activo con partes y hechos específicos
            "conflicto_territorial": (
                "Este texto reporta un CONFLICTO TERRITORIAL ACTIVO Y ESPECÍFICO —disputa "
                "legal, enfrentamiento físico, bloqueo de acceso, invasión u ocupación— "
                "sobre tierras o territorios concretos, con partes en disputa identificadas "
                "y hechos documentados"
            ),
        }

        self.keywords_eventos = {
            "desplazamiento_forzado": [
                "desplazamiento", "desplazados", "desplazar", "expulsión",
                "expulsados", "expulsar", "éxodo", "migración forzada"
            ],
            "reasentamiento": [
                "reasentamiento", "reasentar", "reubicación", "reubicar",
                "traslado", "trasladar", "realojar"
            ],
            "protesta_social": [
                "protesta", "protestas", "protestar", "manifestación",
                "manifestaciones", "bloqueo", "bloquear", "paro", "movilización"
            ],
            "amenaza_intimidacion": [
                "amenaza", "amenazas", "amenazar", "intimidación", "intimidar",
                "hostigamiento", "hostigar", "violencia", "agresión"
            ],
            "consulta_previa": [
                "consulta previa", "consulta", "consentimiento", "socialización",
                "derecho a la consulta", "sin consulta", "omisión de consulta"
            ],
            "audiencia_publica": [
                "audiencia pública", "audiencia", "reunión informativa",
                "socialización del proyecto", "sin audiencia"
            ],
            "taller_participativo": [
                "taller", "talleres", "taller participativo", "taller comunitario",
                "sesión de trabajo", "reunión comunitaria", "sin taller"
            ],
            "conflicto_territorial": [
                "conflicto territorial", "conflicto", "disputa", "litigio",
                "pugna territorial", "enfrentamiento"
            ],
        }

        print(f"Configurado con {len(self.eventos_dict)} tipos de eventos")
        print()

    def _calcular_probabilidades_nli(self, premisa: str, hipotesis: str):
        inputs = self.tokenizer_nli(
            premisa, hipotesis,
            truncation=True, max_length=512,
            padding=True, return_tensors="pt"
        ).to(self.device)

        with torch.no_grad():
            outputs = self.modelo_nli(**inputs)
            probabilidades = torch.softmax(outputs.logits, dim=1)[0]

        return {
            "contradiction": probabilidades[self.label_contradiction].item(),
            "neutral":       probabilidades[self.label_neutral].item(),
            "entailment":    probabilidades[self.label_entailment].item(),
        }

    def extraer_nuggets(self, texto: str, evento_tipo: str):
        if pd.isna(texto) or not texto.strip():
            return []

        doc = self.nlp(texto)
        keywords = self.keywords_eventos.get(evento_tipo, [])
        nuggets = []

        for sent in doc.sents:
            sent_text_lower = sent.text.lower()
            trigger_encontrado = None
            for keyword in keywords:
                if keyword in sent_text_lower:
                    trigger_encontrado = keyword
                    break

            if not trigger_encontrado:
                continue

            agente, verbo, objetos, locacion = None, None, [], None

            for token in sent:
                if token.dep_ == "nsubj":
                    agente = " ".join(child.text for child in token.subtree)
                if token.pos_ == "VERB" and token.dep_ in ["ROOT", "conj"]:
                    verbo = token.lemma_
                if token.dep_ in ["dobj", "obj", "iobj"]:
                    objetos.append(token.text)
                if token.dep_ == "nmod" and token.text.lower() in ["en", "de", "por"]:
                    for child in token.children:
                        if child.pos_ == "PROPN":
                            locacion = child.text

            nuggets.append({
                "oracion":     sent.text.strip(),
                "agente":      agente,
                "verbo":       verbo,
                "trigger":     trigger_encontrado,
                "objetos":     objetos if objetos else None,
                "locacion":    locacion,
                "inicio_char": sent.start_char,
                "fin_char":    sent.end_char,
            })

        return nuggets

    def detectar_eventos(self, texto: str,
                         umbral_entailment: float = 0.5,
                         umbral_neutral_max: float = 0.4,
                         extraer_nuggets: bool = True):
        if pd.isna(texto) or not texto.strip():
            return []

        eventos_detectados = []

        for nombre_corto, hipotesis in self.eventos_dict.items():
            probs = self._calcular_probabilidades_nli(texto, hipotesis)

            if (probs["entailment"] >= umbral_entailment and
                    probs["neutral"] <= umbral_neutral_max):

                evento = {
                    "evento": nombre_corto,
                    "probabilidad_entailment":    float(probs["entailment"]),
                    "probabilidad_neutral":       float(probs["neutral"]),
                    "probabilidad_contradiction": float(probs["contradiction"]),
                }

                if extraer_nuggets:
                    nuggets = self.extraer_nuggets(texto, nombre_corto)
                    evento["nuggets"]     = nuggets
                    evento["num_nuggets"] = len(nuggets)

                eventos_detectados.append(evento)

        eventos_detectados.sort(key=lambda x: x["probabilidad_entailment"], reverse=True)
        return eventos_detectados

    def detectar_en_dataframe(self, df: pd.DataFrame,
                              columna_texto: str = 'texto',
                              umbral_entailment: float = 0.5,
                              umbral_neutral_max: float = 0.4,
                              extraer_nuggets: bool = True):
        print("DETECCIÓN DE EVENTOS")
        print(f"Total de textos: {len(df)}")
        print(f"Umbral entailment: {umbral_entailment}")
        print(f"Umbral neutral máx: {umbral_neutral_max}")
        print()

        resultados = []

        for idx, texto in enumerate(df[columna_texto]):
            if pd.isna(texto):
                resultados.append([])
            else:
                eventos = self.detectar_eventos(
                    texto,
                    umbral_entailment=umbral_entailment,
                    umbral_neutral_max=umbral_neutral_max,
                    extraer_nuggets=extraer_nuggets,
                )
                resultados.append(eventos)

            if (idx + 1) % 10 == 0 or (idx + 1) == len(df):
                print(f"   Procesados: {idx + 1}/{len(df)}")

        print()
        # Columnas numéricas (score 0.0–1.0) — probabilidad de entailment
        # 0.0 = no detectado, >0 = intensidad del evento
        print("   Creando columnas de score numérico...")
        for nombre_corto in self.eventos_dict.keys():
            df[nombre_corto] = [
                next(
                    (ev["probabilidad_entailment"] for ev in resultado
                     if ev["evento"] == nombre_corto),
                    0.0
                )
                for resultado in resultados
            ]

        print("Columnas creadas")
        print()

        return df
```

## **DetectorPosturas**

```python
class DetectorPosturas:
    """
    Detector de posturas y sentimientos en textos.

    Combina:
    1. Análisis de sentimiento (BETO)
    2. Detección de posturas vía NLI
    3. Extracción de nuggets con SpaCy

    Cambios respecto a versión anterior:
    - oposicion_proyecto fusionado en apoyo_proyecto:
        la hipótesis ahora captura el NIVEL DE ACUERDO/DESACUERDO (no solo oposición),
        permitiendo medir polarización como indicador de riesgo.
        En el pipeline, la columna resultante se llama 'nivel_acuerdo_proyecto'.
    - demanda_derechos: reformulada → intensidad de reclamo de derechos
    - propuesta_alternativa: reformulada → señal de déficit de participación efectiva
    - exigencia_participacion: reformulada → intensidad de déficit de participación

    Dimensiones asignadas:
    - nivel_acuerdo_proyecto   → Dim 3: Vulneración socioeconómica (REFORMULAR/FUSIONAR)
    - demanda_derechos         → Dim 5: Derechos humanos y conflicto (REFORMULAR)
    - denuncia_violacion       → Dim 5: Derechos humanos y conflicto (MANTENER)
    - propuesta_alternativa    → Dim 1: Gobernanza y riesgo institucional (REFORMULAR)
    - llamado_dialogo          → Dim 1: Gobernanza — FUSIONAR en pipeline
    - defensa_territorio       → Dim 4: Vulnerabilidad territorial (REFORMULAR)
    - exigencia_participacion  → Dim 1: Gobernanza (REFORMULAR)
    - sentimiento              → transversal — insumo de intensidad para todas las dimensiones
    """

    def __init__(self,
                 modelo_nli="MoritzLaurer/mDeBERTa-v3-base-xnli-multilingual-nli-2mil7",
                 modelo_sentiment="finiteautomata/beto-sentiment-analysis",
                 modelo_spacy="es_core_news_lg",
                 modelo_nli_preloaded=None,
                 tokenizer_nli_preloaded=None):
        print("\n▄ INICIALIZANDO DETECTOR DE POSTURAS")

        # 1. CARGAR MODELO DE SENTIMENT
        print(f"\n[1/3] Cargando modelo de Sentiment...")
        print(f"   Modelo: {modelo_sentiment}")

        self.sentiment_pipeline = transformers_pipeline(
            "sentiment-analysis",
            model=modelo_sentiment,
            device=0 if torch.cuda.is_available() else -1
        )

        self.sentiment_mapping = {
            "POS": "positivo",
            "NEG": "negativo",
            "NEU": "neutral",
        }

        print(f"\u2705 Sentiment pipeline cargado")

        # 2. CARGAR O USAR MODELO NLI COMPARTIDO
        if modelo_nli_preloaded is not None and tokenizer_nli_preloaded is not None:
            print(f"\n[2/3] Usando modelo NLI compartido...")
            self.modelo_nli = modelo_nli_preloaded
            self.tokenizer_nli = tokenizer_nli_preloaded
            self.device = next(self.modelo_nli.parameters()).device
            print(f"\u2705 Modelo NLI compartido conectado")
        else:
            print(f"\n[2/3] Cargando modelo NLI...")
            self.tokenizer_nli = AutoTokenizer.from_pretrained(modelo_nli)
            self.modelo_nli = AutoModelForSequenceClassification.from_pretrained(modelo_nli)
            self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
            self.modelo_nli.to(self.device)
            self.modelo_nli.eval()
            print(f"\u2705 Modelo NLI cargado en: {self.device}")

        self.label_entailment = 0
        self.label_neutral = 1
        self.label_contradiction = 2

        # 3. CARGAR SPACY
        print(f"\n[3/3] Cargando SpaCy para nuggets...")
        try:
            self.nlp = spacy.load(modelo_spacy)
        except OSError:
            import subprocess
            subprocess.run(["python", "-m", "spacy", "download", modelo_spacy])
            self.nlp = spacy.load(modelo_spacy)
        print(f"\u2705 SpaCy cargado")

        # 4. DEFINIR POSTURAS
        # NOTA: oposicion_proyecto se fusionó en apoyo_proyecto →
        #   la hipótesis ahora mide nivel de acuerdo O desacuerdo,
        #   la columna resultante se llama 'nivel_acuerdo_proyecto'.
        #   En el pipeline se interpreta como indicador de polarización/riesgo.
        self.posturas_dict = {
            # TIGHTENED: fusión apoyo+oposición — requiere proyecto nombrado + postura explícita
            "nivel_acuerdo_proyecto": (
                "Este texto expresa explícitamente apoyo o rechazo a UN PROYECTO O INICIATIVA "
                "ESPEC1FICO identificado por nombre o descripción concreta, con una postura "
                "clara de actores sociales o comunitarios frente a ese proyecto particular"
            ),
            # Reformulada: captura intensidad del reclamo, no solo mención
            "demanda_derechos": (
                "Este texto expresa demandas urgentes, exigencias o reclamos intensos "
                "de derechos vulnerados por parte de comunidades o grupos afectados"
            ),
            # TIGHTENED: requiere denuncia formal con actores identificados
            "denuncia_violacion": (
                "Este texto reporta una DENUNCIA FORMAL O EXPL1CITA —tutela, queja ante "
                "autoridad, comunicado oficial, acción legal— de violaciones de derechos "
                "humanos, abusos o incumplimientos graves de acuerdos, presentada por "
                "comunidades, organizaciones o defensores identificados"
            ),
            # Reformulada: señal de que la participación efectiva es insuficiente
            "propuesta_alternativa": (
                "Este texto propone alternativas o soluciones diferentes porque los "
                "canales de participación existentes son insuficientes o han fallado"
            ),
            # Se fusionará en pipeline con participacion_comunitaria, taller, audiencia
            "llamado_dialogo": (
                "Este texto hace un llamado al diálogo, negociación o búsqueda de "
                "acuerdos ante una situación de conflicto o exclusión"
            ),
            # Reformulada: énfasis en conflictividad por defensa
            "defensa_territorio": (
                "Este texto expresa resistencia activa, conflictividad o reivindicación "
                "del territorio, recursos naturales o medio ambiente frente a una amenaza"
            ),
            # Reformulada: déficit percibido de participación
            "exigencia_participacion": (
                "Este texto exige participación, consulta o inclusión porque las "
                "comunidades han sido excluidas de decisiones que las afectan"
            ),
        }

        self.posturas_cortas = list(self.posturas_dict.keys())

        print(f"\n\u2705 Configuración completa:")
        print(f"   - {len(self.posturas_dict)} tipos de posturas")
        print(f"   - 'oposicion_proyecto' fusionado en 'nivel_acuerdo_proyecto'")
        print(f"   - Análisis de sentimiento: BETO (español)")
        print(f"   - Extracción de nuggets: SpaCy")
        print()

    def _analizar_sentimiento(self, texto: str) -> Dict:
        if pd.isna(texto) or not texto.strip():
            return {"sentimiento": "neutral", "confianza": 0.0, "intensidad": "baja"}

        try:
            resultado = self.sentiment_pipeline(texto[:512])[0]
            label_original = resultado['label']
            sentimiento = self.sentiment_mapping.get(label_original, "neutral")
            confianza = resultado['score']

            if confianza >= 0.8:
                intensidad = "alta"
            elif confianza >= 0.6:
                intensidad = "media"
            else:
                intensidad = "baja"

            return {"sentimiento": sentimiento, "confianza": confianza, "intensidad": intensidad}

        except Exception as e:
            print(f"\u26a0 Error en análisis de sentimiento: {e}")
            return {"sentimiento": "neutral", "confianza": 0.0, "intensidad": "baja"}

    def _calcular_probabilidades_nli(self, premisa: str, hipotesis: str) -> Dict[str, float]:
        inputs = self.tokenizer_nli(
            premisa, hipotesis,
            truncation=True, max_length=512,
            padding=True, return_tensors="pt"
        ).to(self.device)

        with torch.no_grad():
            outputs = self.modelo_nli(**inputs)
            probabilidades = torch.softmax(outputs.logits, dim=1)[0]

        return {
            "contradiction": probabilidades[self.label_contradiction].item(),
            "neutral":       probabilidades[self.label_neutral].item(),
            "entailment":    probabilidades[self.label_entailment].item(),
        }

    def _extraer_nuggets_posturas(self, texto: str, posturas_detectadas: List[str]) -> List[Dict]:
        if not posturas_detectadas or pd.isna(texto):
            return []

        doc = self.nlp(texto)
        nuggets = []

        for sent in doc.sents:
            sent_text = sent.text.strip()

            for postura in posturas_detectadas:
                probs = self._calcular_probabilidades_nli(
                    sent_text,
                    self.posturas_dict[postura]
                )

                if probs["entailment"] >= 0.4:
                    sujeto, verbo, objeto = None, None, None

                    for token in sent:
                        if token.dep_ in ["nsubj", "nsubjpass"] and not sujeto:
                            sujeto = token.text
                        if token.pos_ == "VERB" and not verbo:
                            verbo = token.lemma_
                        if token.dep_ in ["obj", "dobj", "iobj"] and not objeto:
                            objeto = token.text

                    nuggets.append({
                        "postura":       postura,
                        "oracion":       sent_text,
                        "sujeto":        sujeto,
                        "verbo":         verbo,
                        "objeto":        objeto,
                        "confianza_nli": probs["entailment"],
                    })

        return nuggets

    def detectar_posturas(
        self,
        texto: str,
        umbral_entailment: float = 0.5,
        umbral_neutral_max: float = 0.4,
        umbral_sentiment: float = 0.7,
        extraer_nuggets: bool = True
    ) -> Dict:
        if pd.isna(texto) or not texto.strip():
            return {
                "posturas": {k: False for k in self.posturas_cortas},
                "posturas_scores": {k: 0.0 for k in self.posturas_cortas},
                "sentimiento": "neutral",
                "sentimiento_confianza": 0.0,
                "sentimiento_intensidad": "baja",
                "nuggets_posturas": [],
            }

        sentiment_result = self._analizar_sentimiento(texto)

        posturas_detectadas = {}
        posturas_encontradas = []

        # Devuelve también los scores crudos para columnas numéricas
        posturas_scores = {}
        for nombre_corto, hipotesis in self.posturas_dict.items():
            probs = self._calcular_probabilidades_nli(texto, hipotesis)
            score = float(probs["entailment"])
            posturas_scores[nombre_corto] = score
            if (
                score >= umbral_entailment
                and probs["neutral"] <= umbral_neutral_max
            ):
                posturas_detectadas[nombre_corto] = True
                posturas_encontradas.append(nombre_corto)
            else:
                posturas_detectadas[nombre_corto] = False

        nuggets = []
        if extraer_nuggets and posturas_encontradas:
            nuggets = self._extraer_nuggets_posturas(texto, posturas_encontradas)

        return {
            "posturas": posturas_detectadas,
            "posturas_scores": posturas_scores,
            "sentimiento": sentiment_result["sentimiento"],
            "sentimiento_confianza": sentiment_result["confianza"],
            "sentimiento_intensidad": sentiment_result["intensidad"],
            "nuggets_posturas": nuggets,
        }

    def clasificar_dataframe(
        self,
        df: pd.DataFrame,
        columna_texto: str = 'texto',
        umbral_entailment: float = 0.5,
        umbral_neutral_max: float = 0.4,
        umbral_sentiment: float = 0.7,
        extraer_nuggets: bool = True
    ) -> pd.DataFrame:
        print("▄ CLASIFICACI2N DE POSTURAS Y SENTIMIENTO")
        print(f"Total de textos: {len(df)}")
        print(f"Umbral entailment: {umbral_entailment}")
        print(f"Umbral neutral m1x: {umbral_neutral_max}")
        print(f"Umbral sentiment: {umbral_sentiment}")
        print(f"Extraer nuggets: {extraer_nuggets}")
        print()

        resultados = []

        for idx, texto in enumerate(df[columna_texto]):
            if pd.isna(texto):
                resultados.append({
                    "posturas": {k: False for k in self.posturas_cortas},
                    "posturas_scores": {k: 0.0 for k in self.posturas_cortas}, # Added this line
                    "sentimiento": "neutral",
                    "sentimiento_confianza": 0.0,
                    "sentimiento_intensidad": "baja",
                    "nuggets_posturas": [],
                })
            else:
                resultado = self.detectar_posturas(
                    texto,
                    umbral_entailment=umbral_entailment,
                    umbral_neutral_max=umbral_neutral_max,
                    umbral_sentiment=umbral_sentiment,
                    extraer_nuggets=extraer_nuggets,
                )
                resultados.append(resultado)

            if (idx + 1) % 10 == 0 or (idx + 1) == len(df):
                print(f"   Procesados: {idx + 1}/{len(df)}")

        print()

        # Columnas numéricas de scores (0.0–1.0) — no binario
        print("   Creando columnas de score numérico de posturas...")
        for nombre_corto in self.posturas_cortas:
            df[nombre_corto] = [
                r["posturas_scores"].get(nombre_corto, 0.0)
                for r in resultados
            ]

        # Columnas de sentimiento — confianza numérica + etiqueta
        print("   Creando columnas de sentimiento...")
        df['sentimiento']            = [r["sentimiento"] for r in resultados]
        df['sentimiento_confianza']  = [r["sentimiento_confianza"] for r in resultados]
        df['sentimiento_intensidad'] = [r["sentimiento_intensidad"] for r in resultados]

        print("✅ Columnas creadas")
        print()

        return df
```

```python
class ClasificadorActores:
    """
    Clasificador de actores e indicadores socio-territoriales.

    Combina:
    1. NER (Transformers): Extrae entidades (personas, organizaciones, lugares)
    2. NLI: Detecta indicadores contextuales
    3. Diccionarios: Categoriza entidades en contexto colombiano

    Cambios respecto a versión anterior:
    - Hipótesis NLI reformuladas en lógica de riesgo/déficit/brecha
    - grupos_etnicos: reformulado → exposición diferencial (MOVER A OTROS en pipeline)
    - equidad_inclusion: reformulado → brechas de inclusión y exclusión social
    - movimientos_sociales: reformulado → intensidad de conflictividad social
    - participacion_economica_local: reformulado → déficit de participación económica
    - transparencia_contractual: reformulado → opacidad contractual y regulatoria
    - respeto_territorios: reformulado → afectación a territorios y medios de vida
    - zonas_proteccion_alimentaria: MOVER A OTROS en pipeline (variable descriptiva)
    - Entidades NER: grupos_etnicos, organizaciones, lideres, instituciones,
      actores_economicos → MOVER A OTROS en pipeline
      grupos_armados_entidades → MANTENER en Dim 5

    Dimensiones asignadas (indicadores NLI):
    - equidad_inclusion              → Dim 3: Vulneración socioeconómica (REFORMULAR)
    - grupos_etnicos                 → Otros (MOVER A OTROS en pipeline)
    - movimientos_sociales           → Dim 3: Vulneración socioeconómica (REFORMULAR)
    - grupos_poblacionales_afectados → Dim 3: Vulneración socioeconómica (MANTENER)
    - participacion_economica_local  → Dim 3: Vulneración socioeconómica (REFORMULAR)
    - transparencia_contractual      → Dim 1: Gobernanza (MANTENER reformulado)
    - zonas_proteccion_alimentaria   → Otros (MOVER A OTROS en pipeline)
    - respeto_territorios            → Dim 4: Vulnerabilidad territorial (REFORMULAR)
    - presencia_grupos_armados       → Dim 5: Derechos humanos y conflicto (MANTENER)
    - desaparicion_lideres           → Dim 5: Derechos humanos y conflicto (MANTENER)
    """

    def __init__(self,
                 modelo_ner="dccuchile/bert-base-spanish-wwm-cased-finetuned-ner",
                 modelo_nli="MoritzLaurer/mDeBERTa-v3-base-xnli-multilingual-nli-2mil7",
                 modelo_nli_preloaded=None,
                 tokenizer_nli_preloaded=None):
        print("\n👥 INICIALIZANDO CLASIFICADOR DE ACTORES")

        # 1. CARGAR MODELO NER
        print(f"\n[1/2] Cargando modelo NER...")
        self.ner_pipeline = transformers_pipeline(
            "ner",
            model=modelo_ner,
            aggregation_strategy="simple",
            device=0 if torch.cuda.is_available() else -1
        )
        print(f"✅ NER pipeline cargado")

        # 2. CARGAR O USAR MODELO NLI COMPARTIDO
        if modelo_nli_preloaded is not None and tokenizer_nli_preloaded is not None:
            print(f"\n[2/2] Usando modelo NLI compartido...")
            self.modelo_nli = modelo_nli_preloaded
            self.tokenizer_nli = tokenizer_nli_preloaded
            self.device = next(self.modelo_nli.parameters()).device
            print(f"✅ Modelo NLI compartido conectado")
        else:
            print(f"\n[2/2] Cargando modelo NLI...")
            self.tokenizer_nli = AutoTokenizer.from_pretrained(modelo_nli)
            self.modelo_nli = AutoModelForSequenceClassification.from_pretrained(modelo_nli)
            self.device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
            self.modelo_nli.to(self.device)
            self.modelo_nli.eval()
            print(f"✅ Modelo NLI cargado en: {self.device}")

        self.label_entailment = 0
        self.label_neutral = 1
        self.label_contradiction = 2

        # 3. INDICADORES NLI — hipótesis reformuladas en lógica de riesgo/déficit
        self.indicadores_dict = {
            # TIGHTENED: requiere brecha concreta documentada, no solo mención de grupos
            "equidad_inclusion": (
                "Este texto describe BRECHAS CONCRETAS DE EXCLUSIÓN —datos, casos o "
                "situaciones específicas— que demuestran desigualdad sistemática en acceso "
                "a servicios, derechos u oportunidades para mujeres, jóvenes, niños, "
                "adultos mayores o grupos étnicos en un territorio determinado"
            ),
            # Variable descriptiva → MOVER A OTROS en pipeline
            "grupos_etnicos": (
                "Este texto menciona exposición diferencial o afectación específica "
                "de pueblos indígenas, comunidades afrodescendientes, raizales, "
                "palenqueros o grupos étnicos"
            ),
            # TIGHTENED: requiere movilización organizada activa con actores identificados
            "movimientos_sociales": (
                "Este texto reporta MOVILIZACIÓN ORGANIZADA Y ACTIVA de movimientos "
                "sociales, colectivos o comunidades —manifestaciones, paros, bloqueos, "
                "plantones— en oposición directa a decisiones institucionales o empresariales "
                "que los afectan, con actores colectivos identificados"
            ),
            # TIGHTENED: requiere población nombrada/cuantificada + causa + magnitud del impacto
            "grupos_poblacionales_afectados": (
                "Este texto identifica EXPLÍCITAMENTE poblaciones concretas —nombradas o "
                "cuantificadas— que han sido afectadas de forma diferenciada y severa por "
                "una causa específica identificada en el texto, describiendo el tipo y "
                "magnitud del impacto sobre esa población"
            ),
            # TIGHTENED: requiere exclusión económica explícita con evidencia concreta
            "participacion_economica_local": (
                "Este texto reporta EXCLUSIÓN EXPLÍCITA de comunidades locales de beneficios "
                "económicos de proyectos o actividades en su territorio —sin empleo local, "
                "sin contratos locales, sin distribución de utilidades— con evidencia "
                "concreta de esa exclusión"
            ),
            # TIGHTENED: requiere irregularidad concreta + entidad verificadora formal
            "transparencia_contractual": (
                "Este texto reporta IRREGULARIDADES CONCRETAS Y VERIFICABLES —sobrecostos, "
                "documentación falsa, licitaciones amañadas, contratos sin supervisión— "
                "en procesos contractuales o de licenciamiento específicos, señalados por "
                "una entidad de control o investigación formal"
            ),
            # Variable descriptiva → MOVER A OTROS en pipeline
            "zonas_proteccion_alimentaria": (
                "Este texto menciona áreas agrícolas, zonas de cultivo, seguridad "
                "alimentaria o soberanía alimentaria"
            ),
            # TIGHTENED: requiere pérdida concreta y medible de medios de vida
            "respeto_territorios": (
                "Este texto reporta PÉRDIDA O DAÑO CONCRETO Y MEDIBLE a territorios, "
                "cultivos, fuentes de agua, prácticas tradicionales o medios de vida "
                "—pesca impedida, cultivos destruidos, acceso bloqueado, sitios profanados— "
                "sufrido por comunidades específicamente identificadas"
            ),
            # Mantener
            "presencia_grupos_armados": (
                "Este texto menciona grupos armados ilegales, actores armados, "
                "guerrilla, paramilitares o violencia armada"
            ),
            # Mantener — indicador crítico
            "desaparicion_lideres": (
                "Este texto menciona amenazas, desapariciones, asesinatos, homicidios "
                "o violencia contra líderes sociales, defensores de derechos humanos "
                "o activistas"
            ),
        }

        # 4. DICCIONARIOS DE REFERENCIA — sin cambios (contexto colombiano)
        self.grupos_etnicos_ref = {
            "wayúu", "wayuu", "guajiro", "arhuaco", "ika", "kogui", "kogi", "wiwa", "kankuamo",
            "emberá", "embera", "wounaan", "zenú", "zenu", "sinú", "sinu",
            "nasa", "páez", "paez", "misak", "guambiano", "inga", "kamëntsá", "kamentsa",
            "pasto", "awá", "awa", "eperara", "tule", "kuna",
            "sikuani", "guahibo", "piapoco", "curripaco", "puinave", "tucano", "desano",
            "cubeo", "guanano", "piratapuyo", "siriano", "carijona", "yukpa", "barí", "bari",
            "u'wa", "uwa", "tunebo", "muisca", "pijao", "yanacona", "coconuco",
            "guambiano", "totoró", "totoro", "quillacinga", "pastos",
            "indígena", "indigena", "pueblo indígena", "pueblo indigena",
            "comunidad indígena", "comunidad indigena", "resguardo indígena", "resguardo indigena",
            "cabildo indígena", "cabildo indigena", "autoridad indígena", "autoridad indigena",
            "territorio ancestral", "territorio indígena", "territorio indigena",
            "afrodescendiente", "afrocolombiano", "afro", "negro", "negra",
            "raizal", "palenquero", "palenquera", "cimarrón", "cimarron",
            "comunidad negra", "comunidad afro", "consejo comunitario",
            "pueblo raizal", "san basilio de palenque", "palenque",
            "chocó", "pacífico colombiano", "pacifico colombiano",
            "onic", "organización nacional indígena", "organizacion nacional indigena",
            "aico", "cric", "consejo regional indígena del cauca", "opiac",
            "gonawindúa", "gonawindua", "confederación indígena tayrona",
        }

        self.grupos_armados_ref = {
            "eln", "ejército de liberación nacional", "ejercito de liberacion nacional",
            "farc", "farc-ep", "fuerzas armadas revolucionarias de colombia",
            "disidencias", "disidencias de las farc", "disidencias farc",
            "epl", "ejército popular de liberación", "ejercito popular de liberacion",
            "m-19", "movimiento 19 de abril",
            "auc", "autodefensas unidas de colombia",
            "agc", "autodefensas gaitanistas de colombia", "clan del golfo",
            "los urabeños", "urabenos", "los rastrojos", "las águilas negras", "aguilas negras",
            "los caparrapos", "caparros", "los puntilleros",
            "paramilitar", "paramilitares", "autodefensas",
            "grupo armado", "grupos armados", "actor armado", "actores armados",
            "grupo armado ilegal", "grupos armados ilegales", "gai",
            "grupo armado organizado", "grupos armados organizados", "gao",
            "guerrilla", "guerrillero", "insurgente", "insurgencia",
            "narcotráfico", "narcotrafico", "narcotraficante", "cartel",
        }

        self.organizaciones_ref = {
            "onic", "aico", "cric", "opiac", "gonawindúa", "gonawindua", "cit",
            "anuc", "asociación nacional de usuarios campesinos",
            "coordinador nacional agrario", "cna", "dignidad agropecuaria",
            "fensuagro", "federación nacional sindical unitaria agropecuaria",
            "marcha patriótica", "marcha patriotica", "congreso de los pueblos",
            "cumbre agraria", "minga social", "minga", "guardia indígena", "guardia indigena",
            "guardia cimarrona", "proceso de comunidades negras", "pcn",
            "comisión colombiana de juristas", "ccj",
            "consultoría para los derechos humanos y el desplazamiento", "codhes",
            "cinep", "centro de investigación y educación popular",
            "defensoría del pueblo", "defensoria del pueblo",
            "cut", "central unitaria de trabajadores", "cgt",
            "fecode", "federación colombiana de educadores",
            "uso", "unión sindical obrera", "union sindical obrera",
            "oit", "organización internacional del trabajo", "onu", "naciones unidas",
            "acnur", "cruz roja", "amnistía internacional", "amnistia internacional",
            "human rights watch", "oxfam", "greenpeace",
            "organización social", "organizacion social", "organización comunitaria",
            "colectivo", "movimiento social", "asociación", "asociacion",
            "fundación", "fundacion", "veeduría ciudadana", "veeduria ciudadana",
            "junta de acción comunal", "junta de accion comunal", "jac",
            "consejo comunitario", "cabildo", "resguardo",
        }

        self.lideres_ref = {
            "líder social", "lider social", "lideresa social",
            "líder comunitario", "lider comunitario", "lideresa comunitaria",
            "líder indígena", "lider indigena", "lideresa indígena", "lideresa indigena",
            "líder campesino", "lider campesino", "lideresa campesina",
            "líder ambiental", "lider ambiental", "lideresa ambiental",
            "defensor de derechos humanos", "defensora de derechos humanos",
            "defensor ambiental", "defensora ambiental",
            "activista", "activista ambiental", "activista social",
            "sindicalista", "comunero", "comunera",
        }

        self.instituciones_ref = {
            "gobierno nacional", "presidencia", "presidente", "vicepresidencia",
            "ministerio", "ministro", "ministra",
            "ministerio de ambiente", "minambiente", "ministerio de minas", "minminas",
            "ministerio del interior", "ministerio de hacienda", "ministerio de agricultura",
            "ministerio de defensa", "ministerio de justicia",
            "anla", "autoridad nacional de licencias ambientales",
            "car", "corporación autónoma regional", "corporacion autonoma regional",
            "parques nacionales", "parques nacionales naturales", "pnn",
            "ideam", "instituto de hidrología",
            "contraloría", "contraloria", "procuraduría", "procuraduria",
            "defensoría del pueblo", "defensoria del pueblo",
            "fiscalía", "fiscalia", "personería", "personeria",
            "alcaldía", "alcaldia", "alcalde", "alcaldesa",
            "gobernación", "gobernacion", "gobernador", "gobernadora",
            "concejo municipal", "asamblea departamental",
            "ejército nacional", "ejercito nacional", "armada nacional",
            "policía nacional", "policia nacional", "esmad", "fuerzas militares",
            "gobierno", "estado", "autoridad", "institución pública",
            "entidad gubernamental", "sector público",
        }

        self.actores_economicos_ref = {
            "empresa minera", "empresa petrolera", "empresa energética", "empresa energetica",
            "multinacional", "transnacional", "sector privado", "sector empresarial",
            "industria extractiva", "industria minera", "industria petrolera",
            "empresa", "compañía", "compania", "corporación", "corporacion",
            "consorcio", "contratista", "operador", "inversionista", "inversor",
            "gremio empresarial", "cámara de comercio", "camara de comercio",
            "ecopetrol", "drummond", "cerrejón", "cerrejon", "anglogold ashanti",
            "continental gold", "gran colombia gold", "isagen", "epm",
            "empresas públicas de medellín", "empresas publicas de medellin",
        }

        print(f"\n✅ Configuración completa:")
        print(f"   - {len(self.indicadores_dict)} indicadores contextuales (hipótesis reformuladas)")
        print(f"   - Diccionarios de referencia sin cambios")
        print()

    def _calcular_probabilidades_nli(self, premisa: str, hipotesis: str) -> Dict[str, float]:
        inputs = self.tokenizer_nli(
            premisa, hipotesis,
            truncation=True, max_length=512,
            padding=True, return_tensors="pt"
        ).to(self.device)

        with torch.no_grad():
            outputs = self.modelo_nli(**inputs)
            probabilidades = torch.softmax(outputs.logits, dim=1)[0]

        return {
            "contradiction": probabilidades[self.label_contradiction].item(),
            "neutral":       probabilidades[self.label_neutral].item(),
            "entailment":    probabilidades[self.label_entailment].item(),
        }

    def extraer_entidades_ner(self, texto: str) -> List[Dict]:
        if pd.isna(texto) or not texto.strip():
            return []
        try:
            return self.ner_pipeline(texto[:512])
        except Exception as e:
            print(f"⚠️ Error en NER: {e}")
            return []

    def categorizar_entidades(self, texto: str, entidades_ner: List[Dict]) -> Dict[str, Set[str]]:
        texto_lower = texto.lower()

        categorias = {
            "grupos_etnicos_entidades":     set(),
            "grupos_armados_entidades":     set(),
            "organizaciones_entidades":     set(),
            "lideres_entidades":            set(),
            "instituciones_entidades":      set(),
            "actores_economicos_entidades": set(),
        }

        for entidad in entidades_ner:
            palabra = entidad['word'].lower().strip()
            if any(ref in palabra for ref in self.grupos_etnicos_ref):
                categorias["grupos_etnicos_entidades"].add(entidad['word'])
            if any(ref in palabra for ref in self.grupos_armados_ref):
                categorias["grupos_armados_entidades"].add(entidad['word'])
            if any(ref in palabra for ref in self.organizaciones_ref):
                categorias["organizaciones_entidades"].add(entidad['word'])
            if any(ref in palabra for ref in self.lideres_ref):
                categorias["lideres_entidades"].add(entidad['word'])
            if any(ref in palabra for ref in self.instituciones_ref):
                categorias["instituciones_entidades"].add(entidad['word'])
            if any(ref in palabra for ref in self.actores_economicos_ref):
                categorias["actores_economicos_entidades"].add(entidad['word'])

        for ref in self.grupos_etnicos_ref:
            if ref in texto_lower:
                match = re.search(rf'\b{re.escape(ref)}\b', texto_lower)
                if match:
                    categorias["grupos_etnicos_entidades"].add(ref.title())
        for ref in self.grupos_armados_ref:
            if ref in texto_lower:
                match = re.search(rf'\b{re.escape(ref)}\b', texto_lower)
                if match:
                    categorias["grupos_armados_entidades"].add(
                        ref.upper() if len(ref) <= 5 else ref.title()
                    )
        for ref in self.organizaciones_ref:
            if ref in texto_lower:
                match = re.search(rf'\b{re.escape(ref)}\b', texto_lower)
                if match:
                    categorias["organizaciones_entidades"].add(
                        ref.upper() if len(ref) <= 5 else ref.title()
                    )
        for ref in self.lideres_ref:
            if ref in texto_lower:
                match = re.search(rf'\b{re.escape(ref)}\b', texto_lower)
                if match:
                    categorias["lideres_entidades"].add(ref.title())
        for ref in self.instituciones_ref:
            if ref in texto_lower:
                match = re.search(rf'\b{re.escape(ref)}\b', texto_lower)
                if match:
                    categorias["instituciones_entidades"].add(ref.title())
        for ref in self.actores_economicos_ref:
            if ref in texto_lower:
                match = re.search(rf'\b{re.escape(ref)}\b', texto_lower)
                if match:
                    categorias["actores_economicos_entidades"].add(ref.title())

        return {k: list(v) if v else [] for k, v in categorias.items()}

    def detectar_indicadores(self,
                             texto: str,
                             umbral_entailment: float = 0.5,
                             umbral_neutral_max: float = 0.4) -> Dict:
        if pd.isna(texto) or not texto.strip():
            return {"indicadores": {}, "entidades": {}}

        indicadores_detectados = {}
        for nombre_corto, hipotesis in self.indicadores_dict.items():
            probs = self._calcular_probabilidades_nli(texto, hipotesis)
            # Guardamos el score crudo — el pipeline decide si supera umbral
            indicadores_detectados[nombre_corto] = float(probs["entailment"])

        entidades_ner = self.extraer_entidades_ner(texto)
        entidades_categorizadas = self.categorizar_entidades(texto, entidades_ner)

        return {"indicadores": indicadores_detectados, "entidades": entidades_categorizadas}

    def clasificar_dataframe(self,
                             df: pd.DataFrame,
                             columna_texto: str = 'texto',
                             umbral_entailment: float = 0.5,
                             umbral_neutral_max: float = 0.4) -> pd.DataFrame:
        print("👥 CLASIFICACIÓN DE ACTORES E INDICADORES")
        print(f"Total de textos: {len(df)}")
        print(f"Umbral entailment: {umbral_entailment}")
        print(f"Umbral neutral máx: {umbral_neutral_max}")
        print()

        resultados = []

        for idx, texto in enumerate(df[columna_texto]):
            if pd.isna(texto):
                resultados.append({
                    "indicadores": {k: False for k in self.indicadores_dict.keys()},
                    "entidades": {
                        "grupos_etnicos_entidades":     [],
                        "grupos_armados_entidades":     [],
                        "organizaciones_entidades":     [],
                        "lideres_entidades":            [],
                        "instituciones_entidades":      [],
                        "actores_economicos_entidades": [],
                    },
                })
            else:
                resultado = self.detectar_indicadores(
                    texto,
                    umbral_entailment=umbral_entailment,
                    umbral_neutral_max=umbral_neutral_max,
                )
                resultados.append(resultado)

            if (idx + 1) % 10 == 0 or (idx + 1) == len(df):
                print(f"   Procesados: {idx + 1}/{len(df)}")

        print()

        # Columnas numéricas para indicadores NLI (score 0.0–1.0)
        print("   Creando columnas de score numérico de indicadores...")
        for nombre_corto in self.indicadores_dict.keys():
            df[nombre_corto] = [
                resultado["indicadores"].get(nombre_corto, 0.0)
                for resultado in resultados
            ]

        # Columnas binarias para entidades NER — no tienen score, solo presencia/ausencia
        print("   Creando columnas binarias de entidades NER...")
        for categoria in [
            "grupos_etnicos_entidades", "grupos_armados_entidades",
            "organizaciones_entidades", "lideres_entidades",
            "instituciones_entidades",  "actores_economicos_entidades",
        ]:
            df[categoria] = [
                len(resultado["entidades"].get(categoria, [])) > 0
                for resultado in resultados
            ]

        df['_entidades_detalle'] = [r["entidades"] for r in resultados]

        print("✅ Columnas creadas")
        print()

        return df
```

#### **Pipeline**

```python
class PipelineClasificacion:
    """
    Orquestador que ejecuta múltiples clasificadores/detectores
    y consolida resultados en columnas unificadas organizadas por dimensión.

    Carga los modelos una sola vez y los comparte entre clasificadores.

    Estructura de dimensiones en la consolidación:
    ─────────────────────────────────────────────────────────────────
    DIM 1: Gobernanza y riesgo institucional
        Indicadores principales:
          · transparencia_contractual       (ClasificadorActores NLI)
          · consulta_previa                 (DetectorEventos)
          · exigencia_participacion         (DetectorPosturas)
          · propuesta_alternativa           (DetectorPosturas — señal de déficit)
        Indicador fusionado (participación y diálogo):
          · deficit_participacion_dialogo    - fusión de:
              participacion_comunitaria     (ClasificadorTemas)
              audiencia_publica             (DetectorEventos)
              taller_participativo          (DetectorEventos)
              llamado_dialogo               (DetectorPosturas)

    DIM 2: Capacidad institucional
          · fortalecimiento_institucional   (ClasificadorTemas)

    DIM 3: Vulneración socioeconómica y comunitaria
          · incentivos_economicos           (ClasificadorTemas — reformulado)
          · protesta_social                 (DetectorEventos)
          · nivel_acuerdo_proyecto          (DetectorPosturas — fusión apoyo/oposición)
          · equidad_inclusion               (ClasificadorActores NLI — reformulado)
          · movimientos_sociales            (ClasificadorActores NLI — reformulado)
          · grupos_poblacionales_afectados  (ClasificadorActores NLI)
          · participacion_economica_local   (ClasificadorActores NLI — reformulado)

    DIM 4: Vulnerabilidad territorial y socioambiental
          · impactos_ambientales            (ClasificadorTemas — reformulado)
          · conflictos_socioambientales     (ClasificadorTemas)
          · reasentamiento                  (DetectorEventos)
          · conflicto_territorial           (DetectorEventos)
          · defensa_territorio              (DetectorPosturas — reformulado)
          · respeto_territorios             (ClasificadorActores NLI — reformulado)

    DIM 5: Derechos humanos y conflicto
          · desplazamiento_forzado          (DetectorEventos)
          · amenaza_intimidacion            (DetectorEventos)
          · denuncia_violacion              (DetectorPosturas)
          · demanda_derechos                (DetectorPosturas — reformulado)
          · presencia_grupos_armados        (ClasificadorActores NLI)
          · desaparicion_lideres            (ClasificadorActores NLI)
          · grupos_armados_entidades        (ClasificadorActores NER)

    OTROS: Contextuales (no entran al radar principal)
          · grupos_etnicos                  (ClasificadorActores NLI)
          · zonas_proteccion_alimentaria    (ClasificadorActores NLI)
          · grupos_etnicos_entidades        (ClasificadorActores NER)
          · organizaciones_entidades        (ClasificadorActores NER)
          · lideres_entidades               (ClasificadorActores NER)
          · instituciones_entidades         (ClasificadorActores NER)
          · actores_economicos_entidades    (ClasificadorActores NER)

    Transversal:
          · sentimiento
          · sentimiento_intensidad
    ─────────────────────────────────────────────────────────────────
    """

    def __init__(self):
        print("INICIALIZANDO PIPELINE DE CLASIFICACIÓN")
        print("=" * 70)

        modelo_nli_nombre = "MoritzLaurer/mDeBERTa-v3-base-xnli-multilingual-nli-2mil7"

        print(f"\nCARGANDO MODELO NLI COMPARTIDO")
        print(f"   Modelo: {modelo_nli_nombre}")

        self.tokenizer_nli_compartido = AutoTokenizer.from_pretrained(modelo_nli_nombre)
        self.modelo_nli_compartido = AutoModelForSequenceClassification.from_pretrained(
            modelo_nli_nombre
        )

        device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        self.modelo_nli_compartido.to(device)
        self.modelo_nli_compartido.eval()

        print(f"Modelo NLI cargado en: {device}")

        print("\n" + "=" * 70)
        print("INSTANCIANDO CLASIFICADORES")
        print("=" * 70)

        print("\n[1/4] Clasificador de Temas Estructurales...")
        self.clasificador_temas = ClasificadorTemas(
            modelo_nli_nombre=modelo_nli_nombre
        )

        print("\n[2/4] Detector de Eventos Puntuales...")
        self.detector_eventos = DetectorEventos(
            modelo_nli_preloaded=self.modelo_nli_compartido,
            tokenizer_nli_preloaded=self.tokenizer_nli_compartido,
        )

        print("\n[3/4] Clasificador de Actores e Indicadores...")
        self.clasificador_actores = ClasificadorActores(
            modelo_nli_preloaded=self.modelo_nli_compartido,
            tokenizer_nli_preloaded=self.tokenizer_nli_compartido,
        )

        print("\n[4/4] Detector de Posturas y Sentimiento...")
        self.detector_posturas = DetectorPosturas(
            modelo_nli_preloaded=self.modelo_nli_compartido,
            tokenizer_nli_preloaded=self.tokenizer_nli_compartido,
        )

        print("\n" + "=" * 70)
        print("✅ PIPELINE INICIALIZADO CORRECTAMENTE")
        print("=" * 70)
        print()

    def procesar_dataframe(self,
                           df: pd.DataFrame,
                           columna_texto: str = 'texto',
                           umbral_temas: float = 0.7,
                           umbral_eventos_entailment: float = 0.7,
                           umbral_eventos_neutral: float = 0.4,
                           umbral_posturas_entailment: float = 0.7,
                           umbral_posturas_neutral: float = 0.4,
                           umbral_actores_entailment: float = 0.7,
                           umbral_actores_neutral: float = 0.4,
                           umbral_sentiment: float = 0.7,
                           extraer_nuggets: bool = True):
        print("\n🔄 PROCESANDO DATAFRAME CON PIPELINE COMPLETO")
        print("=" * 70)
        print(f"Total de textos: {len(df)}\n")

        print("\n[PASO 1/4] Clasificación de Temas")
        print("-" * 70)
        df = self.clasificador_temas.clasificar_dataframe(
            df, columna_texto=columna_texto, umbral=umbral_temas
        )

        print("\n[PASO 2/4] Detección de Eventos Puntuales")
        print("-" * 70)
        df = self.detector_eventos.detectar_en_dataframe(
            df,
            columna_texto=columna_texto,
            umbral_entailment=umbral_eventos_entailment,
            umbral_neutral_max=umbral_eventos_neutral,
            extraer_nuggets=extraer_nuggets,
        )

        print("\n[PASO 3/4] Clasificación de Actores e Indicadores")
        print("-" * 70)
        df = self.clasificador_actores.clasificar_dataframe(
            df,
            columna_texto=columna_texto,
            umbral_entailment=umbral_actores_entailment,
            umbral_neutral_max=umbral_actores_neutral,
        )

        print("\n[PASO 4/4] Detección de Posturas y Sentimiento")
        print("-" * 70)
        df = self.detector_posturas.clasificar_dataframe(
            df,
            columna_texto=columna_texto,
            umbral_entailment=umbral_posturas_entailment,
            umbral_neutral_max=umbral_posturas_neutral,
            umbral_sentiment=umbral_sentiment,
            extraer_nuggets=extraer_nuggets,
        )

        print("\n[CONSOLIDACIÓN] Generando columnas por dimensión")
        print("-" * 70)
        df = self._consolidar_resultados(df)

        print("\n" + "=" * 70)
        print("✅ PROCESAMIENTO COMPLETADO")
        print("=" * 70)
        print()

        return df

    def _consolidar_resultados(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Reorganiza todos los indicadores por dimensión.
        Las columnas individuales son numéricas (0.0–1.0).
        Genera:
          - categorias_detectadas: lista de indicadores que superan UMBRAL_ACTIVO
          - dim1_gobernanza ... dim5_...: listas por dimensión (auditoría)
          - score_dim1 ... score_dim5: promedio de scores 0-1 por dimensión
            → estos son los números para comparar con DANE por municipio
        """

        UMBRAL_ACTIVO = 0.50

        DIM1_INDIVIDUALES = [
            ("transparencia_contractual",  "opacidad_contractual_regulatoria"),
            ("consulta_previa",            "deficit_consulta_previa"),
            ("exigencia_participacion",    "intensidad_deficit_participacion_percibido"),
            ("propuesta_alternativa",      "deficit_canales_participacion_efectiva"),
        ]
        DIM1_FUSION_COLS  = ["participacion_comunitaria","audiencia_publica",
                             "taller_participativo","llamado_dialogo"]
        DIM1_FUSION_LABEL = "deficit_participacion_dialogo_institucional"
        DIM2_INDIVIDUALES = [
            ("fortalecimiento_institucional", "debilidad_institucional_infraestructura"),
        ]
        DIM3_INDIVIDUALES = [
            ("incentivos_economicos",           "concentracion_desigual_beneficios"),
            ("protesta_social",                 "densidad_protesta_social"),
            ("nivel_acuerdo_proyecto",          "nivel_rechazo_polarizacion_proyecto"),
            ("equidad_inclusion",               "brechas_inclusion_exclusion_social"),
            ("movimientos_sociales",            "intensidad_conflictividad_social"),
            ("grupos_poblacionales_afectados",  "poblaciones_diferencialmente_afectadas"),
            ("participacion_economica_local",   "deficit_participacion_economica_local"),
        ]
        DIM4_INDIVIDUALES = [
            ("impactos_ambientales",        "severidad_impactos_socioambientales"),
            ("conflictos_socioambientales", "intensidad_conflictos_socioambientales"),
            ("reasentamiento",              "severidad_reasentamiento_territorial"),
            ("conflicto_territorial",       "intensidad_conflicto_territorial"),
            ("defensa_territorio",          "intensidad_conflictividad_defensa_territorial"),
            ("respeto_territorios",         "afectacion_territorios_medios_de_vida"),
        ]
        DIM5_INDIVIDUALES = [
            ("desplazamiento_forzado",   "intensidad_desplazamiento_forzado"),
            ("amenaza_intimidacion",     "intensidad_amenazas_intimidacion"),
            ("denuncia_violacion",       "denuncia_violacion_derechos"),
            ("demanda_derechos",         "intensidad_reclamo_vulneracion_derechos"),
            ("presencia_grupos_armados", "presencia_densidad_grupos_armados"),
            ("desaparicion_lideres",     "riesgo_critico_lideres_sociales"),
            ("grupos_armados_entidades", "entidades_grupos_armados_detectadas"),
        ]
        OTROS_COLS = [
            ("grupos_etnicos",               "exposicion_diferencial_grupos_etnicos"),
            ("zonas_proteccion_alimentaria", "zonas_proteccion_alimentaria"),
            ("grupos_etnicos_entidades",     "entidades_grupos_etnicos"),
            ("organizaciones_entidades",     "entidades_organizaciones"),
            ("lideres_entidades",            "entidades_lideres"),
            ("instituciones_entidades",      "entidades_instituciones"),
            ("actores_economicos_entidades", "entidades_actores_economicos"),
        ]

        def activo(row, col):
            v = row.get(col, 0.0)
            return bool(v) if isinstance(v, bool) else float(v) >= UMBRAL_ACTIVO

        def consolidar_fila(row):
            cats = []
            for col, label in DIM1_INDIVIDUALES:
                if col in df.columns and activo(row, col):
                    cats.append(f"dim1:{label}")
            if sum(1 for c in DIM1_FUSION_COLS if c in df.columns and activo(row, c)) >= 4:
                cats.append(f"dim1:{DIM1_FUSION_LABEL}")
            for col, label in DIM2_INDIVIDUALES:
                if col in df.columns and activo(row, col):
                    cats.append(f"dim2:{label}")
            for col, label in DIM3_INDIVIDUALES:
                if col in df.columns and activo(row, col):
                    cats.append(f"dim3:{label}")
            for col, label in DIM4_INDIVIDUALES:
                if col in df.columns and activo(row, col):
                    cats.append(f"dim4:{label}")
            for col, label in DIM5_INDIVIDUALES:
                if col in df.columns and activo(row, col):
                    cats.append(f"dim5:{label}")
            if 'sentimiento' in df.columns and pd.notna(row.get('sentimiento')):
                conf = row.get('sentimiento_confianza', '')
                cats.append(f"sentimiento:{row['sentimiento']}:{conf:.2f}"
                             if isinstance(conf, float) else
                             f"sentimiento:{row['sentimiento']}")
            for col, label in OTROS_COLS:
                if col in df.columns and activo(row, col):
                    cats.append(f"otros:{label}")
            entidades_dict = row.get('_entidades_detalle', {}) or {}
            for col, prefijo in {
                'grupos_etnicos_entidades':     'otros:entidades_grupos_etnicos',
                'organizaciones_entidades':     'otros:entidades_organizaciones',
                'lideres_entidades':            'otros:entidades_lideres',
                'instituciones_entidades':      'otros:entidades_instituciones',
                'actores_economicos_entidades': 'otros:entidades_actores_economicos',
                'grupos_armados_entidades':     'dim5:entidades_grupos_armados_detectadas',
            }.items():
                if entidades_dict.get(col):
                    for entidad in entidades_dict[col]:
                        cats.append(f"{prefijo}:{entidad}")
            return cats

        print("   Consolidando indicadores por dimension...")
        df['categorias_detectadas'] = df.apply(consolidar_fila, axis=1)

        def filtrar_dim(cats, prefijo):
            return [c.split(":", 1)[1] for c in cats if c.startswith(prefijo + ":")]

        print("   Creando columnas de lista por dimension...")
        df['dim1_gobernanza']                 = df['categorias_detectadas'].apply(lambda c: filtrar_dim(c, 'dim1'))
        df['dim2_capacidad_institucional']    = df['categorias_detectadas'].apply(lambda c: filtrar_dim(c, 'dim2'))
        df['dim3_vulneracion_socioeconomica'] = df['categorias_detectadas'].apply(lambda c: filtrar_dim(c, 'dim3'))
        df['dim4_vulnerabilidad_territorial'] = df['categorias_detectadas'].apply(lambda c: filtrar_dim(c, 'dim4'))
        df['dim5_derechos_humanos_conflicto'] = df['categorias_detectadas'].apply(lambda c: filtrar_dim(c, 'dim5'))
        df['otros_contextuales']              = df['categorias_detectadas'].apply(lambda c: filtrar_dim(c, 'otros'))

        # Scores numericos por dimension (0-1) para comparar con DANE
        print("   Calculando scores numericos por dimension...")

        def score_dim(row, cols):
            vals = [float(row.get(c, 0.0)) for c in cols
                    if c in df.columns and not isinstance(row.get(c), bool)]
            return round(sum(vals) / len(vals), 4) if vals else 0.0

        cols_dim1 = [c for c, _ in DIM1_INDIVIDUALES] + DIM1_FUSION_COLS
        cols_dim2 = [c for c, _ in DIM2_INDIVIDUALES]
        cols_dim3 = [c for c, _ in DIM3_INDIVIDUALES]
        cols_dim4 = [c for c, _ in DIM4_INDIVIDUALES]
        cols_dim5 = [c for c, _ in DIM5_INDIVIDUALES if c != "grupos_armados_entidades"]

        df['score_dim1_gobernanza']                 = df.apply(lambda r: score_dim(r, cols_dim1), axis=1)
        df['score_dim2_capacidad_institucional']    = df.apply(lambda r: score_dim(r, cols_dim2), axis=1)
        df['score_dim3_vulneracion_socioeconomica'] = df.apply(lambda r: score_dim(r, cols_dim3), axis=1)
        df['score_dim4_vulnerabilidad_territorial'] = df.apply(lambda r: score_dim(r, cols_dim4), axis=1)
        df['score_dim5_derechos_humanos_conflicto'] = df.apply(lambda r: score_dim(r, cols_dim5), axis=1)

        if '_entidades_detalle' in df.columns:
            df.drop(columns=['_entidades_detalle'], inplace=True)

        print("OK: categorias_detectadas, dim1-dim5, score_dim1-score_dim5 creados")
        print()
        return df

    def agregar_por_municipio(self,
                               df: pd.DataFrame,
                               col_municipio: str = 'municipio',
                               col_departamento: str = None) -> pd.DataFrame:
        """
        Agrega scores por municipio para comparar con DANE 2023.

        Por cada municipio calcula:
          - n_articulos: articulos procesados
          - score_dim1..score_dim5: promedio de scores 0-1
          - pct_positivo/negativo/neutral: distribucion de sentimiento

        Args:
            df: DataFrame procesado por procesar_dataframe()
            col_municipio: columna con el nombre del municipio
            col_departamento: columna con el departamento (opcional)

        Returns:
            DataFrame con una fila por municipio y scores 0-1
        """
        score_cols = [
            'score_dim1_gobernanza',
            'score_dim2_capacidad_institucional',
            'score_dim3_vulneracion_socioeconomica',
            'score_dim4_vulnerabilidad_territorial',
            'score_dim5_derechos_humanos_conflicto',
        ]
        group_cols = ([col_departamento, col_municipio]
                      if col_departamento and col_departamento in df.columns
                      else [col_municipio])

        agg = df.groupby(group_cols).agg(
            n_articulos=(col_municipio, 'count'),
            **{col: (col, 'mean') for col in score_cols if col in df.columns}
        ).reset_index()

        if 'sentimiento' in df.columns:
            sent = (df.groupby(group_cols + ['sentimiento'])
                    .size().unstack(fill_value=0).reset_index())
            sent.columns.name = None
            total = sent[['positivo','negativo','neutral']].sum(axis=1).replace(0, 1)
            for s in ['positivo', 'negativo', 'neutral']:
                if s not in sent.columns:
                    sent[s] = 0
                sent[f'pct_{s}'] = (sent[s] / total).round(3)
            agg = agg.merge(
                sent[group_cols + ['pct_positivo','pct_negativo','pct_neutral']],
                on=group_cols, how='left'
            )

        for col in score_cols:
            if col in agg.columns:
                agg[col] = agg[col].round(4)

        return agg
```

#### **Radar**

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from scipy import stats


class CalculadorRadar:
    """
    Calcula un índice tipo Radar por departamento a partir del DataFrame
    procesado por PipelineClasificacion, siguiendo la metodología del
    Radar DANE (PCA por bloque + combinación con pesos + reescalado 0-100).

    Flujo:
        1. Agregar departamento (via columna 'periodico' o columna ya existente)
        2. Calcular tasas de prevalencia por indicador y departamento
        3. Orientar variables (mayor valor = mayor riesgo)
        4. PCA por bloque → 5 scores (0-100) por departamento
        5. Subcomponentes con pesos del Radar oficial
        6. RADAR_propio = 0.6×Corrupción + 0.4×Vulneración
        7. Clasificar por percentiles 25/75

    Uso:
        radar = CalculadorRadar()
        df_radar = radar.calcular(df_procesado)
        resultado = radar.validar_contra_oficial(df_radar, ruta_excel, columna)
    """

    # ── Mapeo periódico :departamento ────────────────────────────────────────
    MAPEO_PERIODICO_DEPARTAMENTO = {
        'El Colombiano':           'Antioquia',
        'El Diario':               'Risaralda',
        'BC Noticias':             'Caldas',
        'El Quindiano':            'Quindío',
        'El País Cali':            'Valle del Cauca',
        'Diario Occidente':        'Valle del Cauca',
        'Diario del Sur':          'Nariño',
        'Diario del Cauca':        'Cauca',
        'Chocó 7 Días':            'Chocó',
        'Llano al Mundo':          'Meta',
        'Diario de Casanare':      'Casanare',
        'La Voz del Cinaruco':     'Arauca',
        'El Morichal':             'Vichada',
        'Mi Putumayo':             'Putumayo',
        'El Tiempo':               'Cundinamarca',
        'La República':            'Cundinamarca',
        'Portafolio':              'Cundinamarca',
        'Publimetro':              'Cundinamarca',
        'Las2Orillas':             'Cundinamarca',
        'El Heraldo':              'Atlántico',
        'El Universal':            'Bolívar',
        'El Pilón':                'Cesar',
        'El Meridiano':            'Córdoba',
        'Vanguardia':              'Santander',
        'Trochando Sin Fronteras': 'Arauca',
        'Enlace Television':       'Santander',
        'Corrillos':               'Santander',
    }

    # ── Definición de bloques para PCA ────────────────────────────────────────
    BLOQUES_PCA = {
        'bloque_A': [  # Corrupción Observable
            'denuncia_violacion',
            'transparencia_contractual',
            'conflicto_territorial',
            'actores_economicos_entidades',
        ],
        'bloque_B': [  # Corrupción Oculta / Violencia
            'presencia_grupos_armados',
            'desaparicion_lideres',
            'amenaza_intimidacion',
            'grupos_armados_entidades',
            'conflictos_socioambientales',
        ],
        'bloque_C': [  # Capacidad Institucional (variables invertidas)
            'fortalecimiento_institucional',
            'llamado_dialogo',
            'instituciones_entidades',
            'propuesta_alternativa',
            'incentivos_economicos',
        ],
        'bloque_D': [  # Vulneración de Derechos
            'desplazamiento_forzado',
            'grupos_etnicos',
            'grupos_poblacionales_afectados',
            'zonas_proteccion_alimentaria',
            'respeto_territorios',
            'equidad_inclusion',
            'grupos_etnicos_entidades',
        ],
        'bloque_E': [  # Participación y Gobernanza (mayormente invertidas)
            'participacion_comunitaria',
            'consulta_previa',
            'audiencia_publica',
            'taller_participativo',
            'exigencia_participacion',
            'movimientos_sociales',
            'organizaciones_entidades',
            'lideres_entidades',
            'nivel_acuerdo_proyecto',   # fusión apoyo/oposición del pipeline
        ],
    }

    # Variables que apuntan hacia RIESGO, NO se invierten
    VARS_RIESGO = {
        'protesta_social', 'amenaza_intimidacion', 'denuncia_violacion',
        'conflicto_territorial', 'desplazamiento_forzado',
        'presencia_grupos_armados', 'desaparicion_lideres',
        'grupos_armados_entidades', 'impactos_ambientales',
        'conflictos_socioambientales', 'demanda_derechos',
        'defensa_territorio', 'exigencia_participacion',
        'grupos_poblacionales_afectados',
    }

    # Variables que apuntan hacia CAPACIDAD  se invierten con (1 - tasa)
    VARS_INVERTIR = {
        'fortalecimiento_institucional', 'llamado_dialogo', 'consulta_previa',
        'audiencia_publica', 'taller_participativo', 'participacion_comunitaria',
        'nivel_acuerdo_proyecto',        # reemplaza apoyo_proyecto
        'propuesta_alternativa', 'incentivos_economicos',
        'participacion_economica_local', 'instituciones_entidades',
    }

    # ── Todas las columnas binarias del pipeline ──────────────────────────────
    COLUMNAS_BINARIAS = [
        # ClasificadorTemas
        'participacion_comunitaria', 'incentivos_economicos',
        'fortalecimiento_institucional', 'impactos_ambientales',
        'conflictos_socioambientales',
        # DetectorEventos
        'desplazamiento_forzado', 'reasentamiento', 'protesta_social',
        'amenaza_intimidacion', 'consulta_previa', 'audiencia_publica',
        'taller_participativo', 'conflicto_territorial',
        # DetectorPosturas
        'nivel_acuerdo_proyecto',   # ← reemplaza apoyo_proyecto + oposicion_proyecto
        'demanda_derechos', 'denuncia_violacion', 'propuesta_alternativa',
        'llamado_dialogo', 'defensa_territorio', 'exigencia_participacion',
        # ClasificadorActores — indicadores
        'equidad_inclusion', 'grupos_etnicos', 'movimientos_sociales',
        'grupos_poblacionales_afectados', 'participacion_economica_local',
        'transparencia_contractual', 'zonas_proteccion_alimentaria',
        'respeto_territorios', 'presencia_grupos_armados', 'desaparicion_lideres',
        # ClasificadorActores — entidades
        'grupos_etnicos_entidades', 'grupos_armados_entidades',
        'organizaciones_entidades', 'lideres_entidades',
        'instituciones_entidades', 'actores_economicos_entidades',
    ]

    # ─────────────────────────────────────────────────────────────────────────

    def calcular(self, df_procesado: pd.DataFrame) -> pd.DataFrame:
        """
        Método principal. Ejecuta las 5 fases y retorna el DataFrame
        con el Radar por departamento.

        Args:
            df_procesado: DataFrame salida de PipelineClasificacion.procesar_dataframe()
                          Debe tener columna 'periodico' o 'departamento'.

        Returns:
            DataFrame con columnas:
            departamento, n_articulos,
            bloque_A, bloque_B, bloque_C, bloque_D, bloque_E,
            corrupcion_score, vulneracion_score,
            radar_propio, categoria_riesgo
        """
        print("=" * 60)
        print("CALCULADOR RADAR — inicio")
        print("=" * 60)

        df = df_procesado.copy()

        df = self._agregar_departamento(df)
        df_tasas   = self._calcular_tasas(df)
        df_bloques = self._aplicar_pca_por_bloque(df_tasas)
        df_sub     = self._calcular_subcomponentes(df_bloques)
        df_radar   = self._calcular_radar_final(df_sub)

        print("\n✅ RADAR CALCULADO")
        print(df_radar[['departamento', 'n_articulos',
                         'corrupcion_score', 'vulneracion_score',
                         'radar_propio', 'categoria_riesgo']].to_string(index=False))

        return df_radar

    # ── Fase 0 ────────────────────────────────────────────────────────────────

    def _agregar_departamento(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Si el DataFrame ya tiene columna 'departamento', la usa directamente.
        Si no, la deriva desde 'periodico' usando MAPEO_PERIODICO_DEPARTAMENTO.
        """
        if 'departamento' in df.columns and df['departamento'].notna().any():
            print("\n[Fase 0] Columna 'departamento' ya existe — se usa directamente.")
            return df

        print("\n[Fase 0] Mapeando 'periodico' → 'departamento'...")
        df['departamento'] = df['periodico'].map(self.MAPEO_PERIODICO_DEPARTAMENTO)

        sin_mapeo = df['departamento'].isna().sum()
        if sin_mapeo > 0:
            periodicos_sin_mapeo = df[df['departamento'].isna()]['periodico'].unique()
            print(f"  Advertencia: {sin_mapeo} artículos sin mapeo.")
            print(f"  Periódicos sin mapeo: {periodicos_sin_mapeo}")
            df = df.dropna(subset=['departamento'])

        print(f"  Departamentos identificados: {sorted(df['departamento'].unique())}")
        return df

    # ── Fase 1 ────────────────────────────────────────────────────────────────

    def _calcular_tasas(self, df: pd.DataFrame) -> pd.DataFrame:
        """
        Agrupa por departamento y calcula tasa de prevalencia por columna
        binaria. Luego orienta las variables (mayor = mayor riesgo).
        """
        print("\n[Fase 1] Calculando tasas de prevalencia por departamento...")

        cols_disponibles = [c for c in self.COLUMNAS_BINARIAS if c in df.columns]
        cols_faltantes   = [c for c in self.COLUMNAS_BINARIAS if c not in df.columns]
        if cols_faltantes:
            print(f"  Columnas no encontradas (se omiten): {cols_faltantes}")

        df_bin = df[['departamento'] + cols_disponibles].copy()
        for col in cols_disponibles:
            df_bin[col] = pd.to_numeric(df_bin[col], errors='coerce').fillna(0)

        # Tasa = proporción de artículos con ese indicador activo
        df_tasas = df_bin.groupby('departamento')[cols_disponibles].mean()
        df_tasas['n_articulos'] = df.groupby('departamento').size()

        print(f"  Departamentos con datos: {len(df_tasas)}")
        print(f"  Artículos por departamento:\n  {df_tasas['n_articulos'].to_dict()}")

        # Orientar variables de capacidad : riesgo
        print("\n  Orientando variables (capacidad → riesgo)...")
        vars_a_invertir = [c for c in cols_disponibles if c in self.VARS_INVERTIR]
        for col in vars_a_invertir:
            df_tasas[col] = 1 - df_tasas[col]
        print(f"  Variables invertidas: {vars_a_invertir}")

        return df_tasas

    # ── Fase 2 ────────────────────────────────────────────────────────────────

    def _aplicar_pca_por_bloque(self, df_tasas: pd.DataFrame) -> pd.DataFrame:
        """
        Aplica PCA(n_components=1) a cada bloque conceptual.
        Si n_departamentos < n_variables del bloque, usa media simple (fallback).
        Reescala cada bloque a 0-100 con min-max.
        """
        print("\n[Fase 2] Aplicando PCA por bloque...")

        n_depts  = len(df_tasas)
        df_bloques = pd.DataFrame(index=df_tasas.index)

        for nombre_bloque, variables in self.BLOQUES_PCA.items():
            vars_disponibles = [v for v in variables if v in df_tasas.columns]

            if not vars_disponibles:
                print(f"  {nombre_bloque}: sin variables disponibles score=0")
                df_bloques[nombre_bloque] = 0.0
                continue

            X = df_tasas[vars_disponibles].values

            if n_depts < len(vars_disponibles):
                print(f"  {nombre_bloque}: fallback a media "
                      f"(n_depts={n_depts} < n_vars={len(vars_disponibles)})")
                scores = X.mean(axis=1)
            else:
                try:
                    scaler   = StandardScaler()
                    X_scaled = scaler.fit_transform(X)
                    pca      = PCA(n_components=1)
                    scores   = pca.fit_transform(X_scaled)[:, 0]
                    varianza = pca.explained_variance_ratio_[0]
                    print(f"  {nombre_bloque}: PCA OK — "
                          f"varianza explicada={varianza:.1%} "
                          f"({len(vars_disponibles)} vars)")
                except Exception as e:
                    print(f"  {nombre_bloque}: error en PCA ({e}) → fallback a media")
                    scores = X.mean(axis=1)

            # Reescalar 0-100 con min-max
            # Con 1 departamento min==max: usar valor crudo escalado al rango teorico
            s_min, s_max = scores.min(), scores.max()
            if s_max > s_min:
                scores_norm = (scores - s_min) / (s_max - s_min) * 100
            elif n_depts == 1:
                scores_norm = np.clip(scores, 0, 1) * 100
            else:
                scores_norm = np.full_like(scores, 50.0, dtype=float)

            df_bloques[nombre_bloque] = scores_norm

        df_bloques['n_articulos'] = df_tasas['n_articulos']
        return df_bloques

    # ── Fase 3 ────────────────────────────────────────────────────────────────

    def _calcular_subcomponentes(self, df_bloques: pd.DataFrame) -> pd.DataFrame:
        """
        Combina los scores de bloque con pesos del Radar oficial DANE:

            corrupcion_score  = 0.8×B + 0.35×A − 0.15×C
            vulneracion_score = 0.5×D + 0.5×E

        Reescala cada subcomponente a 0-100 con min-max.
        """
        print("\n[Fase 3] Calculando subcomponentes con pesos del Radar oficial...")

        df_sub = df_bloques.copy()

        corrupcion_raw = (
            0.80 * df_sub['bloque_B'] +
            0.35 * df_sub['bloque_A'] -
            0.15 * df_sub['bloque_C']
        )
        vulneracion_raw = (
            0.50 * df_sub['bloque_D'] +
            0.50 * df_sub['bloque_E']
        )

        def minmax(serie):
            s_min, s_max = serie.min(), serie.max()
            if s_max > s_min:
                return (serie - s_min) / (s_max - s_min) * 100
            elif len(serie) == 1:
                # Un solo departamento: devolver el valor crudo escalado
                return pd.Series(np.clip(serie.values, 0, 100), index=serie.index)
            return pd.Series(np.full(len(serie), 50.0), index=serie.index)

        df_sub['corrupcion_score']  = minmax(corrupcion_raw)
        df_sub['vulneracion_score'] = minmax(vulneracion_raw)

        print("  corrupcion_score  = 0.8×B + 0.35×A − 0.15×C  (reescalado 0-100)")
        print("  vulneracion_score = 0.5×D + 0.5×E             (reescalado 0-100)")

        return df_sub

    # ── Fase 4 ────────────────────────────────────────────────────────────────

    def _calcular_radar_final(self, df_sub: pd.DataFrame) -> pd.DataFrame:
        """
        Fórmula final:
            RADAR_propio = 0.6×corrupcion_score + 0.4×vulneracion_score

        Clasificación por percentiles 25/75 de la distribución propia.
        """
        print("\n[Fase 4] Calculando Radar final y clasificando por riesgo...")

        df_radar = df_sub.copy()
        df_radar['radar_propio'] = (
            0.60 * df_radar['corrupcion_score'] +
            0.40 * df_radar['vulneracion_score']
        )

        p25 = df_radar['radar_propio'].quantile(0.25)
        p75 = df_radar['radar_propio'].quantile(0.75)

        def clasificar(valor):
            if valor < p25:   return 'Bajo'
            elif valor < p75: return 'Medio'
            else:             return 'Alto'

        df_radar['categoria_riesgo'] = df_radar['radar_propio'].apply(clasificar)

        print(f"  Percentil 25: {p25:.2f}  |  Percentil 75: {p75:.2f}")
        print(f"  Distribución: {df_radar['categoria_riesgo'].value_counts().to_dict()}")

        df_radar = df_radar.reset_index()

        cols_orden = [
            'departamento', 'n_articulos',
            'bloque_A', 'bloque_B', 'bloque_C', 'bloque_D', 'bloque_E',
            'corrupcion_score', 'vulneracion_score',
            'radar_propio', 'categoria_riesgo',
        ]
        cols_presentes = [c for c in cols_orden if c in df_radar.columns]
        return df_radar[cols_presentes]

    # ── Fase 5 ────────────────────────────────────────────────────────────────

    def validar_contra_oficial(self, df_radar_propio: pd.DataFrame,
                                ruta_excel: str,
                                columna_oficial: str,
                                sheet_name: str = '2023',
                                skiprows: int = 3) -> dict:
        """
        Compara el Radar propio con el Radar oficial del DANE.
        Agrega el Excel a nivel departamento (promedio de municipios)
        y calcula correlación de Spearman.

        Args:
            df_radar_propio: DataFrame resultado de calcular()
            ruta_excel:      Ruta al Excel del DANE
            columna_oficial: Nombre de la columna del índice en el Excel
                             Ejemplos:
                               'Radar de Educación'
                               'Radar.Salud'
                               'RADAR.SPD'
                               'Radar.Trabajo'
            sheet_name:      Hoja del Excel. Default: '2023'
            skiprows:        Filas a saltar. Default: 3

        Returns:
            Dict con correlacion, p_valor, n_departamentos,
            interpretacion y df_comparacion
        """
        print("\n[Fase 5] Validando contra Radar oficial DANE...")

        try:
            df_oficial = pd.read_excel(
                ruta_excel, sheet_name=sheet_name, skiprows=skiprows
            ).dropna(how='all')
        except Exception as e:
            print(f"  Error al cargar Excel: {e}")
            return {}

        # Detectar columna de departamento
        col_dept_oficial = None
        for posible in ['Departamento', 'departamento', 'DEPARTAMENTO']:
            if posible in df_oficial.columns:
                col_dept_oficial = posible
                break

        if col_dept_oficial is None:
            print(f"  Error: no se encontró columna 'Departamento' en el Excel.")
            print(f"  Columnas disponibles: {list(df_oficial.columns)}")
            return {}

        if columna_oficial not in df_oficial.columns:
            print(f"  Error: columna '{columna_oficial}' no encontrada.")
            print(f"  Columnas disponibles: {list(df_oficial.columns)}")
            return {}

        # Agregar por departamento — promedio de municipios
        df_ref = (
            df_oficial
            .groupby(col_dept_oficial)[columna_oficial]
            .mean()
            .reset_index()
            .rename(columns={
                col_dept_oficial: 'departamento',
                columna_oficial:  'radar_oficial'
            })
        )

        # Merge
        df_merge = df_radar_propio[['departamento', 'radar_propio']].merge(
            df_ref, on='departamento', how='inner'
        )

        n = len(df_merge)
        print(f"  Departamentos en común: {n}")
        print(f"  {list(df_merge['departamento'])}")

        if n < 3:
            print("  Advertencia: menos de 3 departamentos. "
                  "La correlación no es confiable.")
            return {'n_departamentos': n, 'correlacion': None, 'p_valor': None}

        corr, pval = stats.spearmanr(
            df_merge['radar_propio'],
            df_merge['radar_oficial']
        )

        if pval >= 0.05:
            interp = "No significativa (p ≥ 0.05)"
        elif abs(corr) >= 0.5:
            interp = "Correlación fuerte — el pipeline captura señales reales"
        elif abs(corr) >= 0.3:
            interp = "Correlación moderada — aceptable para proxy basado en NLP"
        else:
            interp = "Correlación débil — revisar indicadores del pipeline"

        print(f"\n  Correlación Spearman: {corr:.4f}")
        print(f"  p-valor:              {pval:.4f}")
        print(f"  Interpretación:       {interp}")
        print(f"\n  Detalle:")
        print(df_merge[['departamento', 'radar_propio', 'radar_oficial']].to_string(index=False))

        return {
            'correlacion':     round(corr, 4),
            'p_valor':         round(pval, 4),
            'n_departamentos': n,
            'interpretacion':  interp,
            'df_comparacion':  df_merge,
        }
```
